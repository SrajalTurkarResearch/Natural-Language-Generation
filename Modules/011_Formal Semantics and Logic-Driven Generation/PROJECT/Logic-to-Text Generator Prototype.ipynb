# Logic-to-Text Generator Prototype

This notebook implements a logic-to-text generator, mapping mathematical expressions, logical rules, or predicates to natural language explanations. Designed with the computational precision of Alan Turing and the formal rigor of Gottlob Frege, it ensures semantic preservation through direct structural mapping.

## Design Overview
- **Inputs**: SymPy strings for math expressions, dictionaries for rules or predicates (inspired by AMR-like structures).
- **Output**: Coherent explanatory text.
- **Features**: Rule-based mapping with potential for entailment checks or lambda calculus extensions.
- **Dependencies**: Requires SymPy (install via `!pip install sympy` if needed).

## Usage
Run each cell to test the generator with provided examples.

```python
import sympy as sp

def logic_to_text(logic_input):
    """
    Converts logical or mathematical input to natural language.
    - Math: SymPy expression string → equation explanation.
    - Rules: Dict {'type': 'rule', 'if': condition, 'then': action} → conditional sentence.
    - Predicates: Dict {'predicate': verb, 'args': [arg1, arg2]} → declarative sentence.
    """
    if isinstance(logic_input, str):  # Math expression
        try:
            expr = sp.sympify(logic_input)
            factored = sp.factor(expr)
            roots = sp.solve(expr)
            explanation = f"The equation {expr} = 0 factors to {factored} = 0."
            if roots:
                explanation += f" It has roots at {', '.join(str(r) for r in roots)}."
            return explanation
        except Exception as e:
            return f"Error in math expression: {e}"
    
    elif isinstance(logic_input, dict):
        if logic_input.get('type') == 'rule':
            return f"If {logic_input['if']}, then {logic_input['then']}."
        elif 'predicate' in logic_input:
            args = logic_input['args']
            if len(args) == 2:
                return f"{args[0].capitalize()} {logic_input['predicate']}s {args[1]}."
            else:
                return "Unsupported predicate arity."
    
    return "Invalid input format."
```

## Example 1: Mathematical Expression

Test with a quadratic equation: `x^2 + 2x + 1`.

```python
math_input = "x**2 + 2*x + 1"
print(logic_to_text(math_input))
```

## Example 2: Logical Rule

Test with a conditional rule: If it rains, then take an umbrella.

```python
rule_input = {'type': 'rule', 'if': 'it rains', 'then': 'take an umbrella'}
print(logic_to_text(rule_input))
```

## Example 3: Predicate

Test with a predicate: John loves Mary.

```python
predicate_input = {'predicate': 'love', 'args': ['John', 'Mary']}
print(logic_to_text(predicate_input))
```
