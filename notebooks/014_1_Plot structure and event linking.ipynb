### 1.2 Plot Structure: The Story Skeleton

**Freytag's Pyramid** (Classic Structure):

| Stage | Purpose | Example | % of Story |
|-------|---------|---------|------------|
| **Exposition** | Setup | "Harry lived in a cupboard" | 20% |
| **Rising Action** | Build tension | "Harry gets Hogwarts letter" | 30% |
| **Climax** | Peak moment | "Harry defeats Voldemort" | 10% |
| **Falling Action** | Resolve | "Harry recovers" | 25% |
| **Resolution** | End | "Harry returns home" | 15% |

````python
# Interactive Freytag's Pyramid Visualization
fig = go.Figure()

stages = ['Exposition', 'Rising Action', 'Climax', 'Falling Action', 'Resolution']
tension = [1, 4, 10, 6, 2]
percent = [20, 30, 10, 25, 15]

fig.add_trace(go.Scatter(x=stages, y=tension, mode='lines+markers+text', 
                        text=[f'{p}%' for p in percent],
                        textposition="top center",
                        name='Tension Level'))

fig.update_layout(title="🗻 Freytag's Pyramid: Story Structure",
                  xaxis_title='Story Stages',
                  yaxis_title='Excitement Level',
                  height=500,
                  showlegend=False)
fig.show()
````

## 1.3 Event Linking: Connecting the Dots

**4 Types of Links**:

| Type | Definition | Example | Symbol |
|------|------------|---------|--------|
| **Causal** | A → causes B | "Rain → Flood" | ➡️ |
| **Temporal** | A before B | "Morning → Afternoon" | ⏰ |
| **Thematic** | Same idea | "All about courage" | 💡 |
| **Spatial** | Same place | "Forest → Cave in forest" | 📍 |

### 1.4 Mathematical Foundation

**Narrative as Graph**: $G = (V, E)$
- $V$ = Events (nodes)
- $E$ = Links (edges with weights 0-1)

**Coherence Score**: $C = \frac{\sum w(e)}{|E|}$

**Path Probability**: $P(path) = \prod P(v_{i+1}|v_i)$

# SECTION 2: STEP-BY-STEP CODE GUIDES

## 2.1 Basic Story Generator (10 minutes)

````python
# STEP 1: Define events by plot stage
events_db = {
    "exposition": ["Alice found a magic book", "Bob got a mysterious letter"],
    "rising_action": ["She read about a hidden cave", "He traveled through dark forest"],
    "climax": ["Alice faced the dragon", "Bob solved the final riddle"],
    "falling_action": ["She escaped with treasure", "He returned home safely"],
    "resolution": ["Alice became a hero", "Bob shared his wisdom"]
}

# STEP 2: Define causal links
links = {
    "exposition": "rising_action",
    "rising_action": "climax",
    "climax": "falling_action",
    "falling_action": "resolution"
}

# STEP 3: Generate story
def generate_story():
    story = []
    stage = "exposition"
    
    while stage:
        event = random.choice(events_db[stage])
        story.append(event)
        stage = links.get(stage)
    
    return " → ".join(story)

# TEST IT!
print("📖 Generated Story:")
print(generate_story())
````

## 2.2 Advanced Graph-Based Generator (15 minutes)

````python
# STEP 1: Create narrative graph
G = nx.DiGraph()

# Add events as nodes
events = [
    "Find map", "Explore cave", "Fight dragon", "Get treasure", "Return home"
]
G.add_nodes_from(events)

# Add weighted edges (causal strength)
edges = [("Find map", "Explore cave", 0.95),
         ("Explore cave", "Fight dragon", 0.85),
         ("Fight dragon", "Get treasure", 0.90),
         ("Get treasure", "Return home", 0.98)]

for u, v, w in edges:
    G.add_edge(u, v, weight=w)

# STEP 2: Calculate coherence
def calculate_coherence(graph):
    weights = [d['weight'] for u, v, d in graph.edges(data=True)]
    if not weights:
        return 0.0
    return sum(weights) / len(weights)

coherence = calculate_coherence(G)
print(f"📊 Coherence Score: {coherence:.3f}")

# STEP 3: Generate path
path = nx.shortest_path(G, source="Find map", target="Return home")
print(f"🗺️  Story Path: {' → '.join(path)}")

# STEP 4: Path probability
prob = np.prod([G[u][v]['weight'] for u, v in zip(path[:-1], path[1:])])
print(f"🎲 Path Probability: {prob:.3f}")
````

# SECTION 3: VISUALIZATIONS & DIAGRAMS

## 3.1 Interactive Narrative Graph

````python
# Visualize the narrative graph
plt.figure(figsize=(12, 8))
pos = nx.spring_layout(G, seed=42)

# Draw nodes
nx.draw_networkx_nodes(G, pos, node_color='lightblue', node_size=2000, alpha=0.9)

# Draw edges with thickness = weight
edges = G.edges(data=True)
for u, v, data in edges:
    nx.draw_networkx_edges(G, pos, edgelist=[(u,v)], 
                          width=data['weight']*5, alpha=0.7,
                          edge_color='gray')

# Labels and weights
nx.draw_networkx_labels(G, pos, font_size=10, font_weight='bold')
edge_labels = {(u,v): f'{d["weight"]:.2f}' for u,v,d in edges}
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=8)

plt.title('🕸️ Narrative Graph: Events & Causal Links', fontsize=16, pad=20)
plt.axis('off')
plt.tight_layout()
plt.show()

print("💡 THICKER lines = STRONGER causal links!")
````

## 3.2 Coherence Analysis Dashboard

````python
# Create 10 random narratives and analyze
narratives = []
coherence_scores = []
lengths = []

for _ in range(10):
    # Find random valid source/target with a path
    valid_path = False
    tries = 0
    max_tries = 50
    path = None
    while not valid_path and tries < max_tries:
        source = random.choice(list(G.nodes))
        target = random.choice(list(G.nodes))
        if source != target:
            try:
                path = nx.shortest_path(G, source=source, target=target)
                valid_path = True
            except nx.NetworkXNoPath:
                valid_path = False
        tries += 1
    if valid_path and path and len(path) > 1:
        prob = np.prod([G[path[i]][path[i+1]]['weight'] 
                       for i in range(len(path)-1)])
        narratives.append(' → '.join(path))
        coherence_scores.append(prob)
        lengths.append(len(path))

# Visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

# Coherence distribution
if coherence_scores:
    sns.histplot(coherence_scores, ax=ax1, bins=5, kde=True, color='skyblue')
    ax1.set_title('📈 Coherence Score Distribution')
    ax1.set_xlabel('Coherence Score')
else:
    ax1.text(0.5,0.5,'No data',va='center',ha='center')

# Length vs Coherence
if coherence_scores and lengths:
    sns.scatterplot(x=lengths, y=coherence_scores, ax=ax2, s=100, color='coral')
    ax2.set_title('📏 Story Length vs Quality')
    ax2.set_xlabel('Number of Events')
    ax2.set_ylabel('Coherence Score')
else:
    ax2.text(0.5,0.5,'No data',va='center',ha='center')

plt.tight_layout()
plt.show()

if coherence_scores:
    print(f"🏆 BEST STORY (highest coherence): {narratives[np.argmax(coherence_scores)]}")
    print(f"📊 AVERAGE COHERENCE: {np.mean(coherence_scores):.3f}")
else:
    print("No valid narratives generated.")
````

# SECTION 4: REAL-WORLD APPLICATIONS

## 4.1 Sports Journalism (Heliograf Style)

````python
# Real sports data
game_events = {
    "exposition": "Lakers vs Warriors, Final Quarter",
    "rising_action": ["LeBron scores 25pts", "Curry hits 3-pointers"],
    "climax": "LeBron game-winning dunk!",
    "falling_action": ["Warriors call timeout", "Fans go wild"],
    "resolution": "Lakers win 115-112!"
}

def sports_report():
    report = []
    for stage in ['exposition', 'rising_action', 'climax', 'falling_action', 'resolution']:
        content = game_events[stage]
        if isinstance(content, list):
            report.append(random.choice(content))
        else:
            report.append(content)
    transitions = ["", "Then", "SUDDENLY", "After", "FINAL SCORE:"]
    return " | ".join([f"{t} {e}".strip() for t, e in zip(transitions, report)])

print("🏀 AUTO-GENERATED SPORTS REPORT:")
print(sports_report())
````

## 4.2 Medical Report Generation

````python
# Patient data → Medical narrative
patient_data = {
    "exposition": "Patient: John Doe, Age 45",
    "rising_action": ["Chest pain reported", "ECG shows abnormality"],
    "climax": "Coronary blockage detected",
    "falling_action": ["Angioplasty performed", "Stent placed successfully"],
    "resolution": "Patient stable, discharged"
}

def medical_report():
    report = f"""
🏥 MEDICAL REPORT
=====================================
"""
    for stage in ['exposition', 'rising_action', 'climax', 'falling_action', 'resolution']:
        content = patient_data[stage]
        if isinstance(content, list):
            report += f"{stage.title()}: {random.choice(content)}\n"
        else:
            report += f"{stage.title()}: {content}\n"
    return report

print(medical_report())
````
