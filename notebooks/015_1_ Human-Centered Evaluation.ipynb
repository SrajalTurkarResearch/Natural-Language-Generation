## 1. Theory & Tutorials: From Fundamentals to Advanced Topics

### 1.1 Fundamentals of NLG
NLG is the process where computers generate human-like text from data. Basic pipeline: Content selection → Discourse planning → Sentence aggregation → Lexicalization → Referring expression → Surface realization.

Example: Input data (weather: sunny, 75°F) → Output: 'Enjoy the sunny day with temperatures at 75°F.'

### 1.2 Traditional vs. Human-Centered Evaluation
Traditional metrics (BLEU, ROUGE) focus on overlap but ignore user context. Human-centered draws from HCI and social sciences for user-focused assessments.

From 2025 updates: Workshops like HEAL@CHI'25 emphasize auditing LLMs with human methods. Improvements since 2015 include LLM-as-judge and error annotations by experts.

### 1.3 Advanced Topics: LLM-Based Evaluation
LLMs as judges approximate human evaluation. Pros: Scalable; Cons: Biases. Taxonomy from surveys: Pairwise, scoring, reference-based.

### 1.4 Human-Centered Frameworks
From EMNLP 2024 tutorial: Integrate social sciences (e.g., surveys) and HCI (e.g., think-aloud). Best practices: Define hypotheses, use multi-item scales.

### 1.5 Readability in Health NLG
Human-Centered Readability Score (HCRS): Dimensions like clarity, trustworthiness. Hybrid auto-human eval.

### 2.2 Computing Traditional Metrics
BLEU measures n-gram overlap. Logic: Precision with brevity penalty.

```python
import nltk
from nltk.translate.bleu_score import sentence_bleu

nltk.download('punkt')

reference = [['The', 'quick', 'brown', 'fox', 'jumps', 'over', 'the', 'lazy', 'dog']]
candidate = ['Quick', 'fox', 'jumps', 'over', 'lazy', 'dog']

score = sentence_bleu(reference, candidate)
print(f'BLEU Score: {score}')
```

### 2.3 LLM-as-Judge Example
Simulate with a simple prompt. In practice, use API.

```python
from transformers import pipeline

judge = pipeline('text-generation', model='gpt2')  # Placeholder; use better model

prompt = "Evaluate if this response is helpful: 'The sky is blue because of Rayleigh scattering.' Helpful means clear and informative. Output: helpful or unhelpful."
result = judge(prompt, max_length=50)[0]['generated_text']
print(result)
```

## 3. Visualizations: Diagrams and Plots

### 3.1 Plotting Metric Correlations
Visualize human vs. auto metric alignment.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

data = pd.DataFrame({
    'BLEU': [0.2, 0.4, 0.6, 0.8],
    'Human Score': [2, 3, 4, 5]
})

sns.scatterplot(data=data, x='BLEU', y='Human Score')
plt.title('Correlation Between BLEU and Human Scores')
plt.show()
```
