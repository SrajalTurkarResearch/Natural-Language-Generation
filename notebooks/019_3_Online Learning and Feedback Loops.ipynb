## 2.3 How Machines Learn: Gradient Descent (Step-by-Step Math)

Imagine guessing your friend's weight:

- Guess: 50 kg
- Real: 70 kg
- **Error** = 20 kg too low

We adjust **a little bit** toward truth.

### Math: Full Calculation

Let **parameter** $w$ = current guess.

- Prediction: $\hat{y} = w$
- Real: $y = 70$
- Loss: $L = (y - \hat{y})^2 = (70 - w)^2$

**Step 1: Derivative (how error changes with $w$)**
$$
\frac{dL}{dw} = 2(70 - w)(-1) = -2(70 - w)
$$

**Step 2: Update Rule**
$$
w_{\text{new}} = w_{\text{old}} - \eta \times \frac{dL}{dw}
$$
where $\eta$ = learning rate (e.g., 0.1)

**Example**:
- $w = 50$
- $\frac{dL}{dw} = -2(70-50) = -40$
- $w_{\text{new}} = 50 - 0.1 \times (-40) = 54$

→ Next guess: 54 kg → **closer!**

**This is how ALL AI learns** — including NLG models.

# Section 3: Natural Language Generation (NLG)

## 3.1 The 3 Steps of NLG

| Step | Name | Job | Example |
|------|------|-----|--------|
| 1 | Content Planning | What to say | Data: 25°C, Sunny → Say "Nice day" |
| 2 | Sentence Planning | How to organize | Weather first, then suggestion |
| 3 | Surface Realization | Grammar | "It's a sunny 25°C day. Go outside!" |

---

## 3.2 Modern NLG: Transformer Models (GPT, Grok, etc.)

**Idea**: Predict **next word** based on previous words.

**Example**:
Input: "The cat sat on the"
→ Model predicts: **mat** (80% chance)

### How It Works
1. **Word → Number Vector** (Embedding)
   - "cat" → [0.1, -0.3, 0.7, ...]
2. **Attention**: Focus on important words
   - When predicting "on", focus on "sat"
3. **Softmax**: Turn scores into probabilities

**Training Loss**: Cross-Entropy
$$
L = -\log P(\text{correct word})
$$

# Section 4: Online Learning — Learning in Real Time

## 4.1 Batch vs Online Learning

| Type | Data | Update | Use Case |
|------|------|--------|--------|
| **Batch** | All data at once | Full retrain | Fixed datasets |
| **Online** | One sample at a time | Immediate update | Live chatbots |

**Analogy**:
- Batch = Study entire book before exam
- Online = Learn as you live

---

## 4.2 Online Learning Math: Stochastic Gradient Descent (SGD)

**Update Rule:**
$$
w_{\text{new}} = w_{\text{old}} - \eta \times \nabla L(\text{one sample})
$$

**Example**:
- Prompt: "The dog is"
- Correct: "happy"
- Model said: "sleepy"
- Loss = 2.3
- Gradient computed → weights updated **immediately**

**No need to retrain on old data!**

# Section 5: Feedback Loops — How AI Improves Itself

## 5.1 What is a Feedback Loop?

**Output → Evaluation → Adjustment → New Output**

**Human Example**:
1. You speak
2. Friend: "Too quiet!"
3. You speak louder
4. Friend: "Perfect!"

---

## 5.2 Types of Feedback

| Type | Effect | Example |
|------|--------|--------|
| **Positive** | Amplifies | Viral post |
| **Negative** | Stabilizes | Thermostat |

**In NLG**: We want **negative feedback** → correct errors.

---

## 5.3 RLHF: Reinforcement Learning from Human Feedback

**Used in ChatGPT, Grok**

### 4-Step Process
1. AI generates 2 responses
2. Human picks better one
3. Train **Reward Model**
4. Update generator using **PPO**

### Reward Model Math
$$
L = -\log(\sigma(r_{\text{better}} - r_{\text{worse}}))
$$

# Section 7: Math Deep Dive (Full Step-by-Step)

## 7.1 Perplexity — Measuring NLG Quality

**Lower = Better**

$$
\text{PPL} = 2^{-\frac{1}{N} \sum \log_2 p(w_i)}
$$

**Example**:
- Sentence: "The cat sat"
- Probabilities: 0.9, 0.8, 0.7
- PPL ≈ 1.25 → **excellent!**

# Section 10: Mini & Major Projects

## Mini Project: Live Feedback Chatbot

```python
# Run this cell
from transformers import GPT2LMHeadModel, GPT2Tokenizer
import torch

tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
model = GPT2LMHeadModel.from_pretrained('gpt2')
optimizer = torch.optim.Adam(model.parameters(), lr=1e-5)

def online_update(prompt, good_response, reward=0.9):
    inputs = tokenizer(prompt, return_tensors='pt')
    labels = tokenizer(good_response, return_tensors='pt')['input_ids']
    outputs = model(**inputs, labels=labels)
    loss = outputs.loss
    # Fix: outputs.logits shape is [batch, seq_len, vocab]; sum over vocab for correct tokens only
    # To use the logits associated with the labels, we align them properly:
    logits = outputs.logits[:, :-1, :]
    labels_cut = labels[:, 1:]
    # Select logits corresponding to the target labels
    logp = torch.log_softmax(logits, dim=-1)
    selected_logp = torch.gather(logp, 2, labels_cut.unsqueeze(-1)).squeeze(-1)
    # Calculate average log-probability of target tokens
    avg_selected_logp = selected_logp.mean()
    total_loss = loss - reward * avg_selected_logp
    total_loss.backward()
    optimizer.step()
    optimizer.zero_grad()
    print(f"Updated with reward: {reward}")

# Try it!
online_update(
    "The dog is",
    "The dog is happy and playful.",
    reward=0.9
)
```
