## Section 1: Theory & Tutorials – From Fundamentals to Advanced

We start with basic concepts, building logically to advanced theories. Each idea is explained with why it exists, mathematical derivations, and simple analogies.

### 1.1 Fundamentals of NLG
Natural Language Generation (NLG) creates human-like text from data. Theory: It mimics human language production – planning content, structuring, and realizing text.

Mathematical foundation: Autoregressive generation, $P(Y|X) = \prod_t P(y_t | y_{<t}, X)$. Derivation: Start with joint probability $P(y_1, y_2, \dots, y_n|X) = P(y_1|X) \cdot P(y_2|y_1,X) \cdot \dots$ (chain rule).

Analogy: Like building a sentence word by word, considering previous words.

### 1.2 Evolution of NLG
From rule-based (1960s) to neural transformers (2017+). In 2025, edge-optimized SLMs like Llama 3.2 use State Space Models (SSMs) for $\mathcal{O}(L)$ complexity vs. $\mathcal{O}(L^2)$ in transformers.

### 1.3 Edge Computing Theory
Edge computing processes data locally. Latency math: $L_{\text{edge}} \approx T_{\text{process}};\ L_{\text{cloud}} = 2 \times T_{\text{transmit}} + T_{\text{process}}$. Example: $T_{\text{transmit}}=50$ ms, $T_{\text{process}}=30$ ms $\rightarrow$ Edge saves $100$ ms.

2025 advancements: Neuromorphic chips (e.g., Intel Loihi 2) reduce energy 15–300× for edge AI.

### 1.4 Advanced Topics: Model Compression
Pruning: Remove weights $|w| < \tau$. Sparsity $s = \text{pruned}/\text{total}$; Speedup $= \frac{1}{1-s}$. Example: $s=0.7 \Rightarrow 3.33\times$ faster.

Quantization: $q = \mathrm{round}\left(\frac{x-z}{s}\right)$. Example calculation: $\min=-10$, $\max=10$, $\text{bits}=8$, $s=\frac{20}{255} \approx 0.078$, $x=5$ $\rightarrow$ $q \approx 192$.

Distillation: $\text{Loss} = \alpha\, \text{CE} + (1-\alpha)\, \text{KL}$. Derivation: $\mathrm{KL}(p\,||\,q) = \sum p \log\left(\frac{p}{q}\right)$.

## Section 2: Practical Code Guides – Step-by-Step

Hands-on code with explanations. Each step includes logic and potential errors.

```python
# Step 1: Load a small NLG model (DistilGPT2)
from transformers import AutoModelForCausalLM, AutoTokenizer

model_name = "distilgpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# Explanation: DistilGPT2 is a compressed version (distillation) for edge. Logic: Fewer parameters (82M vs. 124M in GPT2) reduce memory.
```

```python
# Step 2: Quantize for edge efficiency
import torch
model_quantized = torch.quantization.quantize_dynamic(model, {torch.nn.Linear}, dtype=torch.qint8)

# Explanation: Reduces from 32-bit float to 8-bit int. Memory savings: 4x. Test inference:
input_text = "The future of edge NLG is"
inputs = tokenizer(input_text, return_tensors="pt")
outputs = model_quantized.generate(inputs["input_ids"], max_length=50)
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```

```python
# Visualization 1: Transformer Attention Plot
import matplotlib.pyplot as plt
import numpy as np

# Simulate attention matrix
attention = np.random.rand(5, 5)
fig, ax = plt.subplots()
im = ax.imshow(attention, cmap='hot')
ax.set_title('Attention Mechanism Visualization')
plt.colorbar(im, ax=ax)
plt.xlabel('Key positions')
plt.ylabel('Query positions')
plt.show()

# Explanation: Heatmap shows how words attend to each other. Hotter = stronger focus.
```

```python
# Visualization 2: Edge vs. Cloud Latency Bar Plot
import matplotlib.pyplot as plt

categories = ['Edge', 'Cloud']
latency = [30, 130]  # ms
fig, ax = plt.subplots()
ax.bar(categories, latency, color=['tab:blue', 'tab:orange'])
ax.set_ylabel('Latency (ms)')
ax.set_title('Latency Comparison')
for i, v in enumerate(latency):
    ax.text(i, v + 3, f'{v} ms', ha='center')
plt.show()
```
