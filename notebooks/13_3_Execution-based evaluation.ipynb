## Section 1: Fundamentals of NLG – Understanding the Basics

### 1.1 What is NLG?
Natural Language Generation (NLG) is when a computer takes raw data—like numbers, tables, or images—and turns it into words or sentences that sound human-written. For example, it can turn `{city: 'Paris', temp: 20}` into 'It’s 20 degrees Celsius in Paris today.' NLG is used for reports, chatbots, or even generating code like SQL queries.

- **Key Idea**: NLG makes data understandable by creating text or instructions.
- **Difference from NLU**: NLG *creates* text; Natural Language Understanding (NLU) *understands* text.
- **Analogy**: NLG is like a chef turning ingredients (data) into a recipe (sentences) anyone can follow.

### 1.2 Why NLG Matters
NLG helps computers communicate with people, making data useful in real life. Imagine Einstein explaining relativity in simple words—that’s NLG turning complex data into clear stories.

### 1.3 History and Evolution
- **1950s**: Alan Turing imagined machines talking like humans.
- **1960s**: Eliza, an early chatbot, made simple replies.
- **1970s**: SHRDLU described virtual worlds (e.g., 'The blue block is on the table').
- **2000s**: Statistical models learned patterns from text.
- **2017**: Transformers (from the 'Attention is All You Need' paper) made NLG powerful, like writing essays or code.
- **Why History?**: Like Darwin studying evolution, knowing NLG’s past helps you avoid old mistakes and innovate.

### 1.4 Math Behind NLG
NLG uses probabilities to pick words. A *language model* calculates the chance of each word based on what came before.

- **Formula**: Probability of a sentence = P(word1) × P(word2|word1) × P(word3|word2, word1) × ...
- **Example**: For 'The cat sleeps':
  - P(The) = 0.7, P(cat|The) = 0.6, P(sleeps|cat) = 0.8.
  - Total = 0.7 × 0.6 × 0.8 = 0.336.
  - Perplexity (how 'surprising' the sentence is) = 2^(-1/N × log(prob)), where N = 3.
  - Calculation: log(0.336) ≈ -1.091, -1/3 × -1.091 ≈ 0.364, 2^0.364 ≈ 1.29 (low is good).

## Section 2: Why Evaluate NLG? The Need for Testing

### 2.1 What is Evaluation?
Evaluation checks if an NLG system produces good text. We look at:
- **Fluency**: Does it sound smooth, like a human wrote it?
- **Accuracy**: Are the facts correct?
- **Usefulness**: Does it help with a task (e.g., answer a question)?

- **Analogy**: Like tasting a cake to see if the recipe worked, not just if it looks pretty.

### 2.2 Types of Evaluation
- **Intrinsic**: Checks the text itself (e.g., grammar).
- **Extrinsic**: Checks if the text helps a task (e.g., booking a ticket).
- **Execution-Based**: Checks if the text can 'run' like a program and give the right result (our focus).

### 2.3 Why It’s Important
Testing NLG is like Newton testing gravity with apples—it proves the system works in the real world. Without it, you might trust a model that gives wrong answers.

### 2.4 Math: Measuring Agreement
When humans judge NLG, we use Cohen’s Kappa to check if they agree:
- Formula: κ = (P_o - P_e) / (1 - P_e), where P_o = % agreement, P_e = % chance agreement.
- Example: 2 judges agree on 8/10 outputs (P_o = 0.8), chance = 0.5.
  - κ = (0.8 - 0.5) / (1 - 0.5) = 0.3 / 0.5 = 0.6 (moderate agreement).

```python
# Calculate Cohen’s Kappa
def cohens_kappa(po, pe):
    return (po - pe) / (1 - pe)

po = 0.8  # Observed agreement
pe = 0.5  # Chance agreement
kappa = cohens_kappa(po, pe)
print(f"Cohen’s Kappa: {kappa:.2f}")
```

## Section 3: Execution-Based Evaluation – The Core Topic

### 3.1 What is Execution-Based Evaluation?
Execution-based evaluation tests NLG by running its output as a program or instruction (e.g., SQL query, Python code) and checking if the result is correct. It’s used when NLG creates 'executable' text, like queries for databases.

- **Example**: Question: 'How many employees in sales?' NLG makes SQL: `SELECT COUNT(*) FROM employees WHERE department='sales';`. Run it, check if it returns the right number.
- **Analogy**: Like baking a cake from an NLG recipe and tasting if it’s chocolate as expected.

### 3.2 Why It’s Special
Unlike other tests that check if text looks good, execution-based checks if it *works*. It’s like Turing testing if a computer’s output solves a problem correctly.

### 3.3 History
- **1980s**: Began with semantic parsing (turning questions into queries).
- **2018**: Spider dataset introduced 10,000 questions across 200 databases, making execution-based eval a standard.

### 3.4 Math: Key Metrics
- **Execution Accuracy (EX)**: % of outputs that run and match the true result.
  - Formula: EX = (Correct / Total) × 100%.
  - Example: 10 queries, 7 correct. EX = (7/10) × 100 = 70%.
- **Valid Efficiency Score (VES)**: % that run correctly and fast.
  - Example: 10 queries, 6 valid, 4 efficient. VES = 4/10 = 40%.
- **Test-Suite Accuracy (TS)**: Average over test cases.
  - Formula: TS = (Matches / Tests).
  - Example: 5 tests, 4 match. TS = 4/5 = 80%.
- **F1 Score**: Balances precision (correct outputs given) and recall (correct outputs not missed).
  - Formula: F1 = 2 × (Precision × Recall) / (Precision + Recall).
  - Example: Generated 5 rows, 3 correct, true has 4. Precision = 3/5 = 0.6, Recall = 3/4 = 0.75, F1 = 2 × (0.6 × 0.75) / (0.6 + 0.75) ≈ 0.667.

```python
# Calculate Execution Accuracy and F1 Score
def execution_accuracy(correct, total):
    return (correct / total) * 100

def f1_score(tp, fp, fn):
    precision = tp / (tp + fp)
    recall = tp / (tp + fn)
    return 2 * (precision * recall) / (precision + recall)

# Test EX
correct = 7
total = 10
ex = execution_accuracy(correct, total)
print(f"Execution Accuracy: {ex:.2f}%")

# Test F1
tp = 3  # True positives
fp = 2  # False positives
fn = 1  # False negatives
f1 = f1_score(tp, fp, fn)
print(f"F1 Score: {f1:.3f}")
```

## Section 4: Practical Code Guide – Building an Execution-Based Evaluator

### 4.1 Setting Up a Test Database
Let’s create a simple SQLite database to test NLG-generated SQL queries.

### 4.2 Generating and Testing SQL Queries
We’ll simulate NLG with a simple function and test its output.

```python
import sqlite3
import pandas as pd

# Create a test database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create employees table
cursor.execute('''
CREATE TABLE employees (
    id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary INTEGER
)
''')

# Insert sample data
cursor.executemany('INSERT INTO employees (name, department, salary) VALUES (?, ?, ?)', [
    ('Alice', 'Sales', 50000),
    ('Bob', 'Sales', 55000),
    ('Charlie', 'IT', 60000),
    ('Diana', 'IT', 65000)
])
conn.commit()

# Simulate NLG: Generate SQL query
def generate_sql(question):
    if 'count employees in sales' in question.lower():
        return "SELECT COUNT(*) FROM employees WHERE department='Sales';"
    return None

# Execute and evaluate
question = "How many employees in sales?"
sql_query = generate_sql(question)
true_result = 2  # Known correct answer

try:
    result = cursor.execute(sql_query).fetchone()[0]
    is_correct = result == true_result
    print(f"Query: {sql_query}")
    print(f"Result: {result}, Correct: {is_correct}")
except sqlite3.Error as e:
    print(f"Error: {e}")

conn.close()
```

- **Task**: Use a subset of the Spider dataset to test a transformer-based NLG model.
- **Steps**:
  1. Download Spider (available online).
  2. Fine-tune a T5 model for text-to-SQL.
  3. Evaluate using EX, VES, and TS.

```python
# Mini Project: Simple Text-to-SQL Evaluator
import sqlite3

# Setup database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()
cursor.execute('CREATE TABLE employees (id INTEGER PRIMARY KEY, name TEXT, department TEXT)')
cursor.executemany('INSERT INTO employees (name, department) VALUES (?, ?)', [
    ('Eve', 'HR'), ('Frank', 'HR'), ('Grace', 'Sales')
])
conn.commit()

# Generate and evaluate
questions = ['Count employees in HR', 'List Sales employees', 'Count all employees']
sql_queries = [
    "SELECT COUNT(*) FROM employees WHERE department='HR';",
    "SELECT name FROM employees WHERE department='Sales';",
    "SELECT COUNT(*) FROM employees;"
]
true_results = [2, ['Grace'], 3]

correct = 0
for i, (sql, true) in enumerate(zip(sql_queries, true_results)):
    try:
        result = cursor.execute(sql).fetchall()
        result = result[0][0] if len(result) == 1 and isinstance(true, int) else [r[0] for r in result]
        if result == true:
            correct += 1
        print(f"Question {i+1}: {questions[i]}, Result: {result}, Correct: {result == true}")
    except sqlite3.Error as e:
        print(f"Error in query {i+1}: {e}")

ex = execution_accuracy(correct, len(questions))
print(f"Execution Accuracy: {ex:.2f}%")

conn.close()
```
