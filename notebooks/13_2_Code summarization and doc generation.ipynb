## 1. Theory & Tutorials: From Basics to Advanced

### 1.1 What Are Code Summarization and Doc Generation?
**Idea**: These are ways to use AI to turn computer code into words people understand.
- **Summarization**: Makes a short sentence about what code does. E.g., Code that adds numbers → “Adds two numbers and gives the total.”
- **Doc Generation**: Makes full notes, like instructions for using code, with parts, results, and examples.

**Why It Matters**: As a scientist, you write code for experiments (e.g., analyzing DNA). Auto-summaries and docs save time, reduce errors, and help others repeat your work—key for sharing discoveries, like Curie’s careful notes on radium.

**Logic**: Code has patterns (like loops or names). AI learns these from examples and writes words, like a translator turning math into a story.

**History**: Started in the 1950s with Turing’s ideas, grew in the 1970s with rule systems, 1990s with number-based methods, and 2010s with brain-like nets. In 2025, big word models (LLMs) rule but struggle with tricky code.<grok:render type='render_inline_citation'><argument name='citation_id'>1</argument></grok:render>

**Analogy**: Code is a recipe in shorthand; NLG is a chef explaining it in full sentences.

### 1.2 Basics of NLP and NLG
**Idea**: NLP (Natural Language Processing) is how computers handle human words. NLG is a part that makes new words.

**NLP Parts** (like a kid learning to talk):
- **Tokenization**: Splits words or code into bits (e.g., `addNumbers` → `add`, `Numbers`).
- **Part-of-Speech Tagging**: Labels words as actions (verbs) or things (nouns). In code: Spots functions vs. variables.
- **Named Entity Recognition (NER)**: Finds special names (e.g., variable `x` in code).
- **Parsing**: Builds a tree of how parts connect, like a code map (Abstract Syntax Tree, AST).

**NLG Steps** (like writing a story):
1. Pick what to say (e.g., code’s main job).
2. Put it in order (summary first, details next).
3. Choose words (e.g., ‘adds’ vs. ‘sums’).
4. Use short terms (e.g., ‘it’ instead of repeating ‘function’).
5. Fix grammar.
6. Smooth the text.

**Analogy**: NLP is a librarian sorting books; NLG is an author writing new ones.

**Math Example**: TF-IDF (Term Frequency-Inverse Document Frequency) finds important words.
- TF = (times word appears) / (total words).
- IDF = log(total items / items with word).
- **Calculation**: Code: `add a b`. Word `add` appears 1 time, total words = 3, so TF = 1/3 ≈ 0.333. Two codes total, both have `add`, so IDF = log(2/2) = 0. TF-IDF = 0.333 * 0 = 0 (means `add` is common, not special).

### 1.3 Code Summarization: Deep Dive
**Idea**: Turns code into a short word explanation.

**How It Works**: AI reads code patterns (e.g., loops mean repeating). It learns from big sets like CodeSearchNet (millions of code-word pairs).

**Ways to Do It**:
- **Rules**: If code has `for`, say “loops through.” Simple but stiff.
- **Stats**: Counts words to guess meaning.
- **Brain Nets**: RNNs (order readers), LSTMs (memory holders).
- **Transformers**: Focus on key parts (attention, from 2017, like Einstein’s big idea).<grok:render type='render_inline_citation'><argument name='citation_id'>2</argument></grok:render>

**Math**: Attention in Transformers: Score = softmax(Q * K^T / √d), picks important bits.
- **Calculation**: Q=[0.1, 0.2], K=[[0.3,0.4], [0.5,0.6]], d=2.
  - Q * K^T = [0.1*0.3 + 0.2*0.4, 0.1*0.5 + 0.2*0.6] = [0.11, 0.17].
  - Divide by √2 ≈ 1.41: [0.078, 0.120].
  - Softmax: e^0.078 ≈ 1.081, e^0.120 ≈ 1.128, sum = 2.209, so [0.489, 0.511].

**Hard Parts**: Unclear names (e.g., `x`), long code forgets. Fixes: Better models, shrink code.<grok:render type='render_inline_citation'><argument name='citation_id'>1</argument></grok:render>

**Analogy**: Code is a puzzle; summarization is the box picture.

### 1.4 Doc Generation: Deep Dive
**Idea**: Makes full notes (docstrings, manuals) from code.

**Parts**:
- What it does.
- Inputs (parameters).
- Outputs (returns).
- Errors (exceptions).
- Sample uses.

**How**: Mixes code tree parsing (AST, like a map) with NLG for words.

**Math**: Cosine Similarity picks doc templates: Cos(A,B) = (A·B) / (|A||B|).
- **Calculation**: Code features A=[1,0,1] (function, no loop, add), template B=[1,0,1]. Cos = (1*1 + 0*0 + 1*1) / (√2 * √2) = 2/2 = 1 (perfect match).

**Analogy**: Summarization is a tweet; doc generation is a blog post.

**2025 Note**: New models understand code context better.<grok:render type='render_inline_citation'><argument name='citation_id'>3</argument></grok:render>

### 2.1 Setting Up Your Tools
**Idea**: Use Python and Hugging Face to try NLG.

**Steps**:

2. Load a model like CodeT5 for code tasks.

**Code**: Try summarizing a simple function.

````python
from transformers import pipeline

# Load summarization model (replace with CodeT5 for real use)
summarizer = pipeline('summarization', model='facebook/bart-large-cnn')

# Example code
code = 'def add(a, b): return a + b'
summary = summarizer(code, max_length=50, min_length=10, do_sample=False)
print(summary[0]['summary_text'])  # Output: Adds two numbers and returns their sum.
````

**Explanation**:
- `pipeline`: Easy tool to use AI models.
- `code`: Simple function to test.

### 2.2 Generating a Docstring
**Idea**: Make full notes for code.

**Code**: Create a docstring for a science function.

````python
def simulate_gravity(mass, height):
    g = 9.81
    return 0.5 * mass * g * height**2

# Manual doc (AI would auto-generate similar)
"""
Calculates potential energy under gravity.

Args:
    mass (float): Mass in kilograms.
    height (float): Height in meters.

Returns:
    float: Energy in Joules.

Example:
    >>> simulate_gravity(1, 10)
    490.5
"""
````

**Explanation**:
- Code computes energy (like in physics experiments).
- Docstring lists inputs, outputs, example.
- **Why**: Makes code clear for others, like a lab report.

## 3. Visualizations

### 3.1 Diagram: NLG Pipeline
**Idea**: Show how NLG turns code into words.

**Picture Description**: Draw boxes: `Code Input` → `Parse (AST)` → `Encoder` → `Decoder` → `Text Output`. Arrows between.

**Code**: Plot a simple flowchart.

````python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch, Rectangle

fig, ax = plt.subplots(figsize=(10, 4))
ax.set_xlim(0, 5)
ax.set_ylim(0, 2)

# Boxes
boxes = [('Code', 0.5, 1), ('Parse', 1.5, 1), ('Encoder', 2.5, 1), ('Decoder', 3.5, 1), ('Text', 4.5, 1)]
for label, x, y in boxes:
    ax.add_patch(Rectangle((x-0.4, y-0.2), 0.8, 0.4, fill=False))
    ax.text(x, y, label, ha='center', va='center')

# Arrows
for i in range(len(boxes)-1):
    ax.add_patch(FancyArrowPatch((boxes[i][1]+0.4, 1), (boxes[i+1][1]-0.4, 1), arrowstyle='->'))

ax.axis('off')
plt.title('NLG Pipeline for Code')
plt.show()
````

**Explanation**: Visual shows steps from code to words, like a factory line.

### 3.2 Plot: Model Accuracy
**Idea**: Show how good models are.

**Code**: Fake data for BLEU scores (measures summary quality).

**Task**: Write and summarize a function for Fibonacci numbers.

**Code**:

````python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

# Expected Summary (try with AI later): "Computes nth Fibonacci number recursively."

# Test
print(fibonacci(5))  # Output: 5
````

### 6.2 Major Project: Doc Generation for Climate Data
**Task**: Generate docs for a climate model function using a dataset.

**Dataset**: Use a simple CSV (e.g., temperature data).

**Code**:

````python
import pandas as pd

def average_temp(data_path):
    df = pd.read_csv(data_path)
    return df['temperature'].mean()

# Manual Docstring
"""
Calculates average temperature from a CSV file.

Args:
    data_path (str): Path to CSV with 'temperature' column.

Returns:
    float: Mean temperature in degrees.

Example:
    >>> average_temp('climate_data.csv')
    22.5
"""

# Fake data for demo
data = pd.DataFrame({'temperature': [20, 22, 25]})
data.to_csv('climate_data.csv', index=False)
print(average_temp('climate_data.csv'))  # Output: 22.333
````

**Steps**:
1. Create CSV with temps.
2. Write function.
3. Make docstring manually.
4. Try AI doc generation (e.g., CodeLlama).
