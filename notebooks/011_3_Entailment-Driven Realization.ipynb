## Section 1: Recap of NLG Realization

### 1.1 Theory: What is Realization in NLG?
Since you know NLG turns non-word inputs (like numbers or ideas) into text, **realization** is the final step where a rough plan of ideas (called a semantic structure, like a list of who does what) becomes a proper sentence with correct grammar, word order, and punctuation. It’s like turning a sketch into a polished painting.

The traditional NLG pipeline includes:
1. **Content Determination**: Pick key facts.
2. **Discourse Planning**: Arrange facts logically (e.g., story order).
3. **Sentence Aggregation**: Group ideas to avoid repetition.
4. **Lexicalization**: Choose words (e.g., “hot” vs. “warm”).
5. **Referring Expression Generation**: Use shortcuts like “it” for “the dog.”
6. **Realization**: Apply grammar rules for final sentences.

Realization ensures text is natural and true to the input. In neural NLG (e.g., GPT, T5), these steps blend, but realization still shapes the final output.

### 1.2 Analogy
Realization is like Tesla’s coil turning raw electricity (ideas) into a glowing light (sentences).

### 1.3 Example
- Input: [Who: Sarah, Action: buy, Object: book, Time: past] → Realization: “Sarah bought the book.”
- Formal style: “Ms. Sarah purchased a book.”

### 1.4 Real-World Applications
- **Voice Assistants**: Siri realizes “Set timer” as “Timer set for 5 minutes.”
- **Legal Documents**: Turns contract outlines into “The parties agree to terms.”
- **Science**: Converts experiment data into “Efficiency increased by 10%.”

### 1.5 Visualization
Draw a tree: Top “Sentence,” left branch “Noun Phrase (Sarah),” right “Verb Phrase (bought book).”

### 1.6 Code: Simple Realization Example
Let’s create a rule-based realization for a basic semantic structure.

```python
# Simple rule-based realization
def realize_sentence(semantic_structure):
    # Input: dict with who, action, object, time
    who = semantic_structure['who']
    action = semantic_structure['action']
    obj = semantic_structure['object']
    time = semantic_structure['time']
    
    # Basic grammar rules
    if time == 'past':
        if action == 'buy':
            action = 'bought'
    return f"{who} {action} the {obj}."

# Example
structure = {'who': 'Sarah', 'action': 'buy', 'object': 'book', 'time': 'past'}
print(realize_sentence(structure))  # Output: Sarah bought the book.
```

### 1.7 Math: Grammar Rule Probability
Realization can use probabilities to pick sentence structures. The CYK parsing algorithm calculates the chance of a sentence’s grammar.
- Formula: P(tree) = multiply chances of each rule.
- Example: Rules: Sentence → Noun Phrase + Verb Phrase (0.5), Noun Phrase → “Sarah” (1.0), Verb Phrase → “bought book” (0.4). P = 0.5 * 1.0 * 0.4 = 0.2.

## Section 2: Textual Entailment (NLI) Basics

### 2.1 Theory: What is Entailment?
**Textual Entailment (TE)** or **Natural Language Inference (NLI)** checks if one sentence (the premise, or starting fact) logically leads to another (the hypothesis, or idea being tested). It’s like asking, “If this fact is true, does this idea follow?” Outcomes:
- **Entailment**: Premise makes hypothesis true.
- **Contradiction**: Premise makes hypothesis false.
- **Neutral**: Premise doesn’t decide.

It can be one-way (premise → hypothesis) or two-way (same meaning). It’s probabilistic, mimicking human reasoning, not strict math logic.

### 2.2 Analogy
Like Turing’s code-breaking: Premise is evidence, hypothesis is the code. If evidence fits, it’s entailment.

### 2.3 Examples
- **SNLI**: Premise: “Kids play soccer.” Hypothesis: “Children play a sport.” → Entailment.
- **ANLI**: Premise: “Some birds fly.” Hypothesis: “All animals fly.” → Contradiction.

### 2.4 Applications
- **Fact-Checking**: Google verifies news claims against sources.
- **Question Answering**: Ensures answers match queries (e.g., “Capital of France?” → “Paris”).
- **Science**: Checks if hypotheses match experiment data.

### 2.5 Code: NLI with Hugging Face
Let’s use RoBERTa to classify entailment.

```python
from transformers import pipeline

# Load NLI model
nli_model = pipeline('text-classification', model='facebook/bart-large-mnli')

# Test pair
premise = "Kids play soccer in the park."
hypothesis = "Children are playing a sport."
input_text = f"{premise} [SEP] {hypothesis}"
result = nli_model(input_text)
print(result)  # Expected: High score for 'entailment'
```

### 2.6 Math: Softmax for NLI
NLI models output scores (logits) for each class, turned into chances via softmax.
- Formula: P(class) = exp(score) / sum(exp(all scores)).
- Example: Logits: Entailment=3, Neutral=1, Contradiction=0.
  - exp(3) ≈ 20.085, exp(1) ≈ 2.718, exp(0) = 1.
  - Sum ≈ 23.803.
  - P(Entailment) = 20.085/23.803 ≈ 0.844 (84.4%).

### 2.7 Visualization
Plot a bar chart of softmax probabilities.

```python
import matplotlib.pyplot as plt
import numpy as np

# Softmax example
logits = [3, 1, 0]
exp_logits = [np.exp(x) for x in logits]
sum_exp = sum(exp_logits)
probs = [x/sum_exp for x in exp_logits]

plt.bar(['Entailment', 'Neutral', 'Contradiction'], probs)
plt.title('NLI Softmax Probabilities')
plt.ylabel('Probability')
plt.show()
```

## Section 3: Why Entailment in NLG?

### 3.1 Theory: Ensuring Faithful Text
**Faithfulness** means generated text matches the input—no made-up facts (hallucinations) or missing details. Entailment checks if:
- Input entails output (no omissions).
- Output entails input (no additions).

Methods:
- **Post-Generation**: Check text after it’s made.
- **During Generation**: Guide word choices with entailment scores.

### 3.2 Analogy
Like Einstein checking if a conclusion follows from evidence.

### 3.3 Examples
- Input: [Sales: Up 10%] → Text: “Sales rose 10%.” → Entailment: Yes.
- Bad: “Sales soared dramatically.” → Neutral (adds “dramatically”).

### 3.4 Applications
- **Summarization**: Ensures news summaries match articles.
- **Science**: Verifies auto-generated reports match data.
- **Machine Reading**: E3 extracts rules with entailment.

### 3.5 Code: Faithfulness Check
Check if generated text entails input.

```python
# Simplified faithfulness check
def check_entailment(premise, hypothesis, model):
    input_text = f"{premise} [SEP] {hypothesis}"
    result = model(input_text)
    # Some models return label in uppercase, so normalize
    label = result[0]['label'].lower()
    return label == 'entailment' and result[0]['score'] > 0.7

premise = "Sales increased by 10%."
hypothesis = "Sales rose 10%."
print(check_entailment(premise, hypothesis, nli_model))  # Expected: True
```

### 3.6 Math: Faithfulness Score
F = (P(entailment | output, input) + P(entailment | input, output)) / 2.
- Example: P(output | input) = 0.9, P(input | output) = 0.8. F = (0.9 + 0.8) / 2 = 0.85.

## Section 4: Core: Entailment-Driven Realization

### 4.1 Theory: Guiding Realization with Entailment
Entailment-driven realization uses NLI to ensure final sentences match the input’s meaning. Instead of just grammar, it picks sentences with the highest entailment scores.
- **Post-Generation**: Generate options, rank by NLI.
- **During Generation**: Use reinforcement learning (reward entailment) or iterative checks.
- **Entailment Trees**: Build logical steps (nodes) where each entails the next, great for complex tasks like question answering.

### 4.2 Analogy
Like Tesla’s wireless power: Entailment sends meaning without loss.

### 4.3 Examples
- Input: [Team A wins, Team B, 2-1]
  - Option 1: “Team A won 2-1.” → High entailment.
  - Option 2: “Team A crushed Team B.” → Neutral.
- **Entailment Tree**: Question: “Did Canada win Olympics?”
  - Top: “Canada won?”
  - Middle: “Most gold medals?”
  - Bottom: “Canada: 10, USA: 9.”

### 4.4 Applications
- **Logical NLG**: Table-to-text with verified statements.
- **Multimodal QA**: Combines images/text (e.g., bridge safety).
- **Explanations**: NILE generates fact-based reasons.
- **Science**: Auto-generates faithful abstracts.

### 4.5 Code: Entailment-Driven Generation
Generate text and select best via NLI.

```python
from transformers import T5ForConditionalGeneration, T5Tokenizer

# Load NLG model
t5_model = T5ForConditionalGeneration.from_pretrained('t5-small')
t5_tokenizer = T5Tokenizer.from_pretrained('t5-small')

# Generate candidates
input_data = "Team A wins, Team B, score 2-1"
prompt = f"Generate text from: {input_data}"
inputs = t5_tokenizer(prompt, return_tensors='pt')
outputs = t5_model.generate(inputs['input_ids'], num_beams=3, num_return_sequences=3)
candidates = [t5_tokenizer.decode(out, skip_special_tokens=True) for out in outputs]

# Rank with NLI
best_candidate = None
best_score = 0
for cand in candidates:
    score = nli_model(f"{input_data} [SEP] {cand}")[0]['score'] if check_entailment(input_data, cand, nli_model) else 0
    if score > best_score:
        best_score = score
        best_candidate = cand
print(f"Best: {best_candidate}")
```

### 4.6 Math: Reinforcement Reward
Reward R = α * Entailment + β * Fluency + γ * Coverage.
- Example: α=0.5, β=0.3, γ=0.2. Scores: 0.9, 0.8, 0.7.
  - R = (0.5 * 0.9) + (0.3 * 0.8) + (0.2 * 0.7) = 0.45 + 0.24 + 0.14 = 0.83.

## Section 7: Major Project: Entailment-Driven NLG System

### 7.1 Objective
Build an NLG system using ToTTo dataset, generating text and verifying with NLI.

### 7.2 Code
Generate and check text.

```python
# Load ToTTo
totto = load_dataset('totto', split='train[:10]')

# Generate and verify
def generate_and_verify(table, t5_model, t5_tokenizer, nli_model):
    prompt = f"Generate text from table: {table['table']}"
    inputs = t5_tokenizer(prompt, return_tensors='pt')
    outputs = t5_model.generate(inputs['input_ids'], num_beams=3, num_return_sequences=3)
    candidates = [t5_tokenizer.decode(out, skip_special_tokens=True) for out in outputs]
    best = None
    best_score = 0
    for x in candidates:
        score = nli_model(f"{table['table']} [SEP] {x}")[0]['score'] if check_entailment(table['table'], x, nli_model) else 0
        if score > best_score:
            best_score = score
            best = x
    return best

table = totto[0]
print(generate_and_verify(table, t5_model, t5_tokenizer, nli_model))
```

## Section 8: Exercises & Solutions

### 8.1 Exercise 1: Realization
Write a function to realize [Who: John, Action: run, Time: present] as “John runs.”

**Solution**:

```python
def realize_exercise(structure):
    # Simple present tense for 'run' (add 's' for 3rd person singular)
    verb = structure['action']
    if structure.get('time', '') == 'present' and structure['who'].lower() not in ['i', 'you', 'we', 'they']:
        if verb.endswith('y') and verb[-2] not in 'aeiou':
            verb = verb[:-1] + 'ies'
        elif verb.endswith(('s', 'sh', 'ch', 'x', 'z')):
            verb = verb + 'es'
        else:
            verb = verb + 's'
    return f"{structure['who']} {verb}."

print(realize_exercise({'who': 'John', 'action': 'run', 'time': 'present'}))  # John runs.
```
