## 2. Core Theory: What is Empathy in AI?

### 2.1 Three Types of Empathy

| Type | Human Version | AI Simulation |
|------|---------------|---------------|
| **Cognitive** | "I understand you're sad" | Detect emotion from text |
| **Affective** | "I feel sad with you" | Use warm, resonant language |
| **Compassionate** | "Let me help" | Offer support, resources |

### 2.2 Challenges in AI Empathy

- **Ambiguity**: "I'm fine" → often means upset
- **Context Dependency**: "I lost" → game or loved one?
- **Cultural Variation**: Direct vs. indirect expression
- **Sincerity Risk**: Over-empathy feels fake

**Analogy**: Teaching a robot to hug — it can learn the motion, but making it *feel* warm takes deep understanding.

## 4. Emotion Detection: From Keywords to Transformers

### 4.1 Level 1: Keyword Matching (Beginner)


````python
def keyword_emotion(text):
    text = text.lower()
    if any(word in text for word in ['happy', 'great', 'excited']):
        return 'joy'
    elif any(word in text for word in ['sad', 'down', 'depressed']):
        return 'sadness'
    elif any(word in text for word in ['angry', 'mad', 'furious']):
        return 'anger'
    else:
        return 'neutral'

# Test
print(keyword_emotion("I'm so happy!"))  # joy
print(keyword_emotion("I failed..."))    # neutral (misses nuance!)
````

### 4.2 Level 2: Pre-trained Transformer Model


````python
from transformers import pipeline

# Load emotion classifier
emotion_pipeline = pipeline(
    "text-classification",
    model="bhadresh-savani/distilbert-base-uncased-emotion",
    return_all_scores=True
)

def detect_emotion_transformers(text):
    results = emotion_pipeline(text)[0]
    return {r['label']: r['score'] for r in results}

# Test
text = "I just lost my job and feel hopeless."
emotions = detect_emotion_transformers(text)
print(emotions)
````

## 5. Context Understanding & Memory

### 5.1 Dialogue History Tracking


````python
class ContextManager:
    def __init__(self, max_history=5):
        self.history = []
        self.max_history = max_history
    
    def add(self, user_msg, bot_msg=None):
        self.history.append({"user": user_msg, "bot": bot_msg})
        if len(self.history) > self.max_history:
            self.history.pop(0)
    
    def get_context(self):
        return "\n".join([
            f"User: {turn['user']}\nBot: {turn['bot']}"
            for turn in self.history if turn['bot'] is not None
        ])

# Demo
context = ContextManager()
context.add("I had a bad day at work")
context.add("My boss yelled at me", "That's awful. Want to talk?")
print(context.get_context())
````

## 6. Response Generation with Empathy

### 6.1 Prompt Engineering for Empathy


````python
from transformers import pipeline

generator = pipeline("text-generation", model="facebook/blenderbot-400M-distill")

def generate_empathetic_response(user_input, emotion, context=""):
    prompt = (
        f"You are an empathetic friend. The user is feeling {emotion}.\n"
        f"Context: {context}\n"
        f"User: {user_input}\n"
        f"Respond with warmth, validation, and support.\n"
        f"Response:"
    )

    output = generator(prompt, max_length=100, truncation=True)[0]['generated_text']
    # Sometimes there is more than one 'Response:', so we split from the first occurrence
    if "Response:" in output:
        return output.split("Response:", 1)[-1].strip()
    else:
        return output.strip()

# Test
response = generate_empathetic_response(
    "I failed my exam", 
    "sadness", 
    "User previously said: 'I've been studying so hard'"
)
print(response)
````

## 7. Mathematics of Empathy Modeling

### 7.1 Softmax for Emotion Probabilities

$$
P(\text{emotion}_i) = \frac{e^{s_i}}{\sum_j e^{s_j}}
$$


````python
import numpy as np

def softmax(scores):
    scores = np.asarray(scores)
    exp_scores = np.exp(scores - np.max(scores))  # stability improvement
    return exp_scores / np.sum(exp_scores)

# Example: model outputs raw scores
scores = np.array([2.0, 1.0, 0.1, -0.5])  # joy, sadness, anger, fear
probs = softmax(scores)

emotions = ['joy', 'sadness', 'anger', 'fear']
for emo, prob in zip(emotions, probs):
    print(f"{emo}: {prob:.3f}")
````

### 7.2 Cross-Entropy Loss for Training

$$
L = -\sum y_i \log(\hat{y}_i)
$$

````python
def cross_entropy_loss(y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    return -np.sum(y_true * np.log(y_pred + 1e-8))

# True: sadness (one-hot), Predicted: softmax output
y_true = np.array([0, 1, 0, 0])
y_pred = softmax(np.array([0.1, 2.5, 0.2, -1.0]))

loss = cross_entropy_loss(y_true, y_pred)
print(f"Loss: {loss:.4f}")
````

## 10. Mini Project: Build Your First Empathetic Chatbot

**Goal**: Create a working chatbot that detects emotion and responds empathetically.


````python
from transformers import pipeline

class EmpatheticChatbot:
    def __init__(self):
        self.emotion_pipe = pipeline("text-classification", 
                                     model="bhadresh-savani/distilbert-base-uncased-emotion")
        self.generator = pipeline("text-generation", 
                                  model="facebook/blenderbot-400M-distill")
        self.context = ContextManager()
    
    def respond(self, user_input):
        # Detect emotion
        emotion = self.emotion_pipe(user_input)[0]['label']
        
        # Generate response
        prompt = f"User feels {emotion}. Be empathetic: {user_input}"
        response = self.generator(prompt, max_length=80)[0]['generated_text']
        
        # Update context
        self.context.add(user_input, response)
        
        if ":" in response:
            return response.split(":", 1)[-1].strip()
        return response.strip()

# Run chatbot
bot = EmpatheticChatbot()
print(bot.respond("I just broke up with my partner"))
````

## 13. Research Directions & Rare Insights

### Rare Insight 1: **Empathy Decay**
- Over long conversations, models **lose emotional alignment**.
- **Fix**: Inject emotion tokens every 3 turns.

### Rare Insight 2: **Cultural Empathy Gap**
- Western datasets → direct empathy.
- Asian datasets → indirect, harmony-focused.
- **Solution**: Train on **multicultural** empathetic corpora.
