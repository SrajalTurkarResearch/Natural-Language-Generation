## Section 2: Practical Code Guides

Step-by-step code with explanations. We'll use Python, NLTK for rules, HuggingFace for neural.

```
# Install if needed (commented as per env)
#!pip install nltk transformers

import nltk
import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer

# Rule-Based Example
def rule_based_nlg(data):
    return f"The temperature in {data['city']} is {data['temp']}°C."

data = {'city': 'Paris', 'temp': 25}
print(rule_based_nlg(data))
```

Explanation: Simple template filling.

```
# Neural Example
tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
model = GPT2LMHeadModel.from_pretrained('gpt2')

input_text = 'Weather in Paris: 25°C. It is'
inputs = tokenizer(input_text, return_tensors='pt')
outputs = model.generate(**inputs)
print(tokenizer.decode(outputs[0]))
```

Explanation: GPT2 generates fluent continuation.

```
# Neurosymbolic: Rule-constrained Neural
def neurosymbolic_nlg(data, model, tokenizer):
    base_text = rule_based_nlg(data)
    inputs = tokenizer(base_text + ' Additional info:', return_tensors='pt')
    outputs = model.generate(**inputs)
    return tokenizer.decode(outputs[0])

print(neurosymbolic_nlg(data, model, tokenizer))
```

Explanation: Rule provides base, neural adds fluency.

```
# Application: Medical Report
def medical_nlg(patient_data):
    # Rule + Neural
    base = f"Patient age: {patient_data['age']}, symptom: {patient_data['symptom']}."
    inputs = tokenizer(base + ' Diagnosis:', return_tensors='pt')
    outputs = model.generate(**inputs)
    return tokenizer.decode(outputs[0])

patient = {'age': 40, 'symptom': 'fever'}
print(medical_nlg(patient))
```

## Section 7: Exercises

1. Implement a rule for cold weather.
2. Train a small neural model on custom data.

Solutions below.

```
# Exercise 1 Solution
def extended_rule(data):
    if data['temp'] < 10:
        return f"It's cold in {data['city']} at {data['temp']}°C."
    else:
        return rule_based_nlg(data)
print(extended_rule({'city': 'Paris', 'temp': 5}))
```
