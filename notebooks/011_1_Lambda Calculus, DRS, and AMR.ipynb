## Section 1: Introduction to NLG and Key Concepts

### 1.1 What is Natural Language Generation (NLG)?
- **Theory**: NLG is how computers turn data (like numbers or facts) into human-readable text, like stories or reports. It’s part of computational linguistics, studying how computers handle language. NLG has three steps:
  1. **Content Determination**: Decide what to say (e.g., “temperature is 100°C” → focus on boiling).
  2. **Sentence Planning**: Organize the structure (e.g., “Water is boiling” vs. “It’s hot”).

- **Semantics**: The meaning behind words, handled by Lambda, DRS, and AMR to ensure logic and avoid AI mistakes (called hallucinations).
- **History**: Started in 1960s with rule-based systems like SHRDLU; now uses neural networks like GPT, but Lambda, DRS, AMR provide a reliable base.
- **Analogy**: NLG is like Tesla’s electricity system: data is raw power, semantic tools are transformers, and text is a glowing bulb.
- **2025 Update**: NLG powers multimodal AIs (text + images), with a market worth billions, per recent reports.

## Section 2: Lambda Calculus – The Engine of Functional Computation

### 2.1 Theory and History
- **What is It?**: Invented by Alonzo Church in 1936, Lambda Calculus is a way to describe computations using functions. Everything is a function: take an input, give an output, no side effects (like pure math).
- **Parts**:
  - **Variable**: A placeholder, like x or y.
  - **Abstraction**: λx.M means a function taking x and doing M (e.g., λx.x is “return x”).
  - **Application**: (M N) means apply function M to input N.
  - **Free vs. Bound**: In λx.x y, x is bound (tied to λ), y is free.
- **History**: Proved computation limits (like Gödel’s theorems) and matches Turing’s machines in power.
- **Analogy**: Like Tesla’s wireless power—functions send results without messy wires (changing variables).

**Visualization**: Sketch a function:
```
λx.          (head: function takes x)
  |
  x+1        (body: what it does)
```

### 2.2 Core Rules and Math
- **Rules**:
  - **Alpha Conversion**: Rename variables to avoid confusion, e.g., λx.x → λy.y.
  - **Beta Reduction**: Substitute, e.g., (λx.x+1) 5 → 5+1 → 6.
  - **Eta Conversion**: Simplify, e.g., λx.(f x) → f if x isn’t free in f.
- **Church-Rosser**: Reductions reach the same result, like physics laws.
- **Church Numerals**: Numbers as functions, e.g., 2 = λf.λx.f (f x) (apply f twice).
- **Math Example**: Add 2 + 1.
  - 2 = λf.λx.f (f x), 1 = λf.λx.f x.
  - Addition: λm.λn.λf.λx.m f (n f x).
  - Step-by-step:
    1. Start: (λm.λn.λf.λx.m f (n f x)) (λf.λx.f (f x)) (λf.λx.f x).
    2. Replace m: λn.λf.λx.(λf.λx.f (f x)) f (n f x) → λn.λf.λx.f (f (n f x)).
    3. Replace n: λf.λx.f (f ((λf.λx.f x) f x)) → λf.λx.f (f (f x)) = 3.

**Code Example**: Simulate lambda in Python.

````python
# Lambda Calculus in Python
# Identity function: λx.x
identity = lambda x: x
print(identity(5))  # Outputs: 5

# Church numerals: 2 = λf.λx.f (f x)
two = lambda f: lambda x: f(f(x))
three = lambda f: lambda x: f(f(f(x)))

# Addition: λm.λn.λf.λx.m f (n f x)
add = lambda m: lambda n: lambda f: lambda x: m(f)(n(f)(x))

# Test: Add 2 + 1
inc = lambda x: x + 1  # Increment function
result = add(two)(three)(inc)(0)  # Apply inc 2+3 times to 0
print(result)  # Outputs: 5
````

**Visualization**: Plot Church numeral applications.
```
y-axis: Result (e.g., 0, 1, 2, ...)
x-axis: Number of applications (f applied n times)
Line: y = x for numeral n
```

````python
import matplotlib.pyplot as plt

def plot_church_numeral(n, label):
    f = lambda x: x + 1
    x = range(5)
    y = [n(f)(0) for _ in x]  # Apply n times
    plt.plot(x, y, label=label)

plot_church_numeral(two, 'Two')
plot_church_numeral(three, 'Three')
plt.xlabel('Applications')
plt.ylabel('Result')
plt.legend()
plt.show()
````

### 2.3 Advanced Concepts
- **Y-Combinator**: For recursion, Y = λf.(λx.f (x x)) (λx.f (x x)). Enables loops, e.g., factorial.
- **Typed Lambda**: Add types (e.g., int → bool) to catch errors; System F for flexible types.
- **Pitfalls**: Avoid infinite loops like (λx.x x) (λx.x x).

**Mini Project**: Implement factorial in Python using lambda recursion.
- Task: Compute 5! = 5*4*3*2*1.
- Steps: Define Y-combinator, write factorial function, test for n=5.

````python
# Y-Combinator for recursion
Y = lambda f: (lambda x: f(lambda z: x(x)(z)))(lambda x: f(lambda z: x(x)(z)))

# Factorial: λf.λn. if n=0 then 1 else n * f(n-1)
factorial = Y(lambda f: lambda n: 1 if n == 0 else n * f(n-1))
print(factorial(5))  # Outputs: 120
````

## Section 3: Discourse Representation Structures (DRS) – Contextual Semantics

### 3.1 Theory and History
- **What is DRS?**: Created by Hans Kamp in 1981, DRS represents sentence meanings in boxes, handling pronouns and time. Part of Discourse Representation Theory for multi-sentence logic.
- **Components**:
  - **Universe**: List of variables (e.g., x for “John”).
  - **Conditions**: Facts like man(x), loves(x,y).
  - **Nested Boxes**: For if-then (⇒), not (¬), or (∨).
- **Analogy**: Like Einstein’s spacetime—context shapes meaning connections.

**Visualization**: Draw a box:
```
[x]        // Universe: x exists
man(x)     // Conditions
smiles(x)
```

### 3.2 Construction and Operations
- **Building DRS**: Merge boxes for new sentences if variables are accessible.
- **Quantifiers**: “Every” (∀) uses ⇒, “a” (∃) adds variables.
- **Math**: Verify truth by mapping variables to a model where conditions hold.
- **Example**: “If a man is happy, he smiles.”
  - DRS: [] ⇒ [x] man(x) happy(x) [] smiles(x).
  - Check: For all x (man and happy), smiles is true.

````python
# Simple DRS Representation in Python
class DRS:
    def __init__(self, universe, conditions):
        self.universe = universe
        self.conditions = conditions

    def __str__(self):
        return f"Universe: {self.universe}\nConditions: {self.conditions}"

# Example: A man smiles
drs = DRS(['x'], ['man(x)', 'smiles(x)'])
print(drs)
````

### 3.3 Advanced Features
- **Tense**: Add event variables (e), e.g., e < now for past.
- **Modals**: Boxes for possibility, e.g., ♦ [ ] P(x).
- **Plurals**: Group variables, e.g., sum(x,y) for collectives.
- **Pitfalls**: Ensure pronouns access correct variables.

### 3.4 Applications and Research Insights
- **NLG Example**: “Scientist invents machine. It revolutionizes energy.” DRS links “it” to machine.
- **Real-World**: Tesla patent descriptions, linking “coil” across sentences.
- **Scientific Case**: Model experiment chains: “If experiment succeeds, theory holds.”
- **Rare Insight**: DRS ensures coherence in multi-turn scientific dialogues.

### 3.5 2025 Updates and Research Frontiers
- **Universal DRT**: Anchors DRS to words for cross-language NLG, e.g., science reports.

## Section 4: Abstract Meaning Representation (AMR) – Graph Semantics

### 4.1 Theory and History
- **What is AMR?**: Developed in 2013, AMR represents sentence meanings as graphs, ignoring grammar. Uses PropBank for verb senses.
- **Components**:
  - **Nodes**: Concepts (e.g., jump-01).
  - **Edges**: Relations (e.g., :ARG0 for agent).
  - **Reentrancy**: Loops for coreference.
- **Analogy**: Like Tesla’s circuit diagrams—nodes are parts, edges are connections.

**Visualization**: Draw a graph:
```
jump-01
/      \
:ARG0  :ARG1
fox    dog
```

````python
import networkx as nx
import matplotlib.pyplot as plt

# Simple AMR Graph
G = nx.DiGraph()
G.add_nodes_from(['jump-01', 'fox', 'dog'])
G.add_edges_from([('jump-01', 'fox', {'label': ':ARG0'}), ('jump-01', 'dog', {'label': ':ARG1'})])

pos = nx.spring_layout(G)
nx.draw(G, pos, with_labels=True)
edge_labels = nx.get_edge_attributes(G, 'label')
nx.draw_networkx_edge_labels(G, pos, edge_labels)
plt.show()
````

### 4.2 Operations and Algorithms
- **Parsing**: Text to graph using rules or neural models.
- **Smatch Metric**: Compare graphs: F1 = 2*(precision*recall)/(precision+recall).
- **Math Example**: Match graphs for “Boy wants apple” vs. “Girl desires fruit.” Nodes match 80%, edges 90%, F1 = 0.848.

### 4.3 Applications and Research Insights
- **NLG Example**: “Quick brown fox jumps over lazy dog.” AMR: (j / jump-01 :ARG0 (f / fox :mod (b / brown) :mod (q / quick)) :ARG1 (d / dog :mod (l / lazy))).
- **Real-World**: Engineering CAD reports: “Bolt secures plate.”
- **Scientific Case**: Chemistry: AMR for “H2O forms from H and O.”
- **Rare Insight**: AMR’s abstraction enables multilingual NLG.

## Section 5: Major Project – NLG System for Scientific Reports

**Goal**: Build an NLG system to generate a physics report from data using Lambda, DRS, and AMR.

**Dataset**: Simple physics data (velocity, mass, acceleration).

**Steps**:
1. **Data**: {mass: 2kg, velocity: 10m/s, acceleration: 5m/s²}.
2. **AMR**: Represent as graph: (move-01 :ARG1 (object :mass 2) :velocity 10 :acceleration 5).
3. **DRS**: Add context: [x,e] object(x) mass(x,2) event(e) move(e,x) velocity(e,10) acceleration(e,5).
4. **Lambda**: Compute force: λm.λa.m*a applied to 2, 5 → 10N.
5. **Generate Text**: “An object of 2kg moves at 10m/s with 5m/s² acceleration, exerting 10N force.”

````python
# NLG System
data = {'mass': 2, 'velocity': 10, 'acceleration': 5}

# Lambda for force
force = lambda m: lambda a: m * a
force_value = force(data['mass'])(data['acceleration'])

# Simple DRS
drs = DRS(['x', 'e'], [f'object(x)', f'mass(x,{data["mass"]}kg)', f'move(e,x)', f'velocity(e,{data["velocity"]}m/s)', f'acceleration(e,{data["acceleration"]}m/s²)'])
print(drs)

# AMR Graph
G = nx.DiGraph()
G.add_nodes_from(['move-01', 'object', 'mass', 'velocity', 'acceleration'])
G.add_edges_from([('move-01', 'object', {'label': ':ARG1'}), ('object', 'mass', {'label': ':mass'}), ('move-01', 'velocity', {'label': ':velocity'}), ('move-01', 'acceleration', {'label': ':acceleration'})])

pos = nx.spring_layout(G)
nx.draw(G, pos, with_labels=True)
nx.draw_networkx_edge_labels(G, pos, nx.get_edge_attributes(G, 'label'))
plt.show()

# Generate Text
text = f"An object of {data['mass']}kg moves at {data['velocity']}m/s with {data['acceleration']}m/s² acceleration, exerting {force_value}N force."
print(text)
````
