## Section 1: Quantization### FundamentalsQuantization reduces numerical precision, e.g., FP32 to INT8, shrinking size and speeding computations.### Advanced TheoryPost-Training (PTQ) vs. Quantization-Aware Training (QAT). Recent: 4-bit quantization for LLMs (2025 advances).Math: Uniform quantization s = (max - min)/(2^b -1), z = round(-min/s), q = clip(round(T/s + z), 0, 2^b-1).

```python
# Example: Simple quantizationimport numpy as npweights = np.array([1.2, 3.7, -0.5])b = 8min_val, max_val = weights.min(), weights.max()s = (max_val - min_val) / (2**b - 1)z = np.round(-min_val / s)q = np.clip(np.round(weights / s + z), 0, 2**b - 1)deq = s * (q - z)print("Original:", weights)print("Quantized:", q)print("Dequantized:", deq)
```

## Section 3: Distillation### FundamentalsTeacher-student knowledge transfer.### Advanced TheoryKD loss: tau^2 KL(p_t || p_s). Recent: Adaptive distillation in SPADE (2025).Math: p = softmax(z/tau).

```python
# Simple distillation lossimport torchimport torch.nn.functional as Fz_t = torch.tensor([2.0, 1.0, 0.0])z_s = torch.tensor([1.5, 1.0, 0.5])tau = 2.0p_t = F.softmax(z_t / tau, dim=0)p_s = F.softmax(z_s / tau, dim=0)kl = F.kl_div(p_s.log(), p_t, reduction='batchmean') * tau**2print("KL Loss:", kl.item())
```
