Math Tutorial: Cosine similarity for embeddings:
$$  \text{Similarity} = \frac{\mathbf{A} \cdot \mathbf{B}}{\|\mathbf{A}\| \|\mathbf{B}\|} $$
Example: Compute similarity between 'happy' and 'joyful' vectors.

```python
import numpy as np

# Example embeddings
happy = np.array([0.5, 0.3, 0.2])
joyful = np.array([0.4, 0.35, 0.15])

# Cosine similarity
similarity = np.dot(happy, joyful) / (np.linalg.norm(happy) * np.linalg.norm(joyful))
print(f'Similarity: {similarity:.3f}')
```


## 3. Practical Code Guides

### 3.1 Basic Affective NLG System
Step-by-step: Use VADER for sentiment, Hugging Face for generation.

```python
# Install if needed (assuming environment has them)
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer
from transformers import pipeline

nltk.download('vader_lexicon')

analyzer = SentimentIntensityAnalyzer()
generator = pipeline('text-generation', model='gpt2')

def generate_affective_text(input_text):
    scores = analyzer.polarity_scores(input_text)
    sentiment = 'positive' if scores['compound'] > 0.3 else 'negative' if scores['compound'] < -0.3 else 'neutral'
    prompt = f'Respond {sentiment}ly: {input_text}'
    return generator(prompt, max_length=50)[0]['generated_text']

print(generate_affective_text('I am sad today.'))
```
