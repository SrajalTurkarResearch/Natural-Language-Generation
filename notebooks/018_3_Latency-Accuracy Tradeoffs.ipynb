## Section 1: Theory & Tutorials

### 1.1 Fundamentals of NLG
Natural Language Generation (NLG) is the process of producing human-readable text from structured or unstructured data using AI models. Start with basics: NLG transforms inputs like numerical data into coherent sentences.

Theory: NLG relies on probabilistic modeling. In autoregressive models, text is generated word-by-word, where each word's probability depends on previous ones. Mathematically, the probability of a sequence y given input x is P(y|x) = ∏ P(y_i | y_{1:i-1}, x).

Logic: This decomposition ensures coherence but introduces sequential computation, a key source of latency.

### 1.2 Accuracy in NLG
Accuracy measures output quality: correctness, fluency, relevance. Metrics include:
- BLEU: N-gram overlap with references. Formula: BLEU = BP * exp(∑ w_n log p_n), where BP is brevity penalty, p_n is precision.
- Perplexity: 2^{-(1/N) ∑ log P(y_i)}, lower is better.

Advanced: Embeddings-based metrics like BERTScore use cosine similarity for semantic accuracy.

### 1.3 Latency in NLG
Latency is the time from input to output. In transformers, it's O(L * n^2 * d) due to attention mechanisms.

### 1.4 The Tradeoff
Larger models improve accuracy via better representations but increase latency. Scaling laws: Accuracy ∝ log(parameters), latency ∝ parameters.

### 1.5 Optimization Techniques
Compression (pruning, quantization), efficient inference (speculative decoding), context management.

Advanced: In RAG, adaptive retrieval reduces latency by 45% with minor accuracy loss.

## Section 2: Practical Code Guides

Here, we implement concepts step-by-step using PyTorch for simplicity.

```python
import torch
import torch.nn as nn
import time
import numpy as np

# Step 1: Simple Autoregressive Model (LSTM for NLG basics)
class SimpleNLG(nn.Module):
    def __init__(self, vocab_size, embed_size, hidden_size):
        super().__init__()
        self.embed = nn.Embedding(vocab_size, embed_size)
        self.lstm = nn.LSTM(embed_size, hidden_size, batch_first=True)
        self.fc = nn.Linear(hidden_size, vocab_size)
    
    def forward(self, x):
        embed = self.embed(x)
        out, _ = self.lstm(embed)
        return self.fc(out)

# Explanation: This model embeds input tokens, processes via LSTM, predicts next token.
# To measure latency and accuracy:
vocab_size = 10000  # Example vocabulary
model_small = SimpleNLG(10000, 128, 256)  # Small model for low latency
model_large = SimpleNLG(10000, 512, 1024)  # Large for high accuracy

# Dummy input (batch 1, seq len 50)
input_tensor = torch.randint(0, vocab_size, (1, 50))

# Measure latency
start = time.time()
_ = model_small(input_tensor)
latency_small = time.time() - start
print(f'Small model latency: {latency_small:.4f} seconds')

start = time.time()
_ = model_large(input_tensor)
latency_large = time.time() - start
print(f'Large model latency: {latency_large:.4f} seconds')

# For accuracy, in practice use perplexity on a dataset. Here, simulate.
# Logic: Larger hidden size captures better patterns, lower perplexity.
```

Explanation: The code above demonstrates model size impact. Small model: faster but potentially higher perplexity. Extend to transformers for advanced.

```python
# Advanced: Using HuggingFace for real NLG (assume transformers installed)
from transformers import GPT2LMHeadModel, GPT2Tokenizer
import time

tokenizer = GPT2Tokenizer.from_pretrained('gpt2')
model = GPT2LMHeadModel.from_pretrained('gpt2')

prompt = 'The cat sat on the'
inputs = tokenizer(prompt, return_tensors='pt')

start = time.time()
outputs = model.generate(inputs['input_ids'], max_length=50)
latency = time.time() - start
print(f'Generation latency: {latency:.4f} seconds')
print(tokenizer.decode(outputs[0], skip_special_tokens=True))

# To trade off: Use gpt2 (small) vs gpt2-large (load accordingly).
```
