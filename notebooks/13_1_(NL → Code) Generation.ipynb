### 1.2 The Evolution of NLP
NLP grew like a tree from a tiny seed:
- **1950s-80s: Rule-Based NLP**: Humans wrote strict rules, like “If a sentence has ‘is,’ it’s a statement.” It was like giving a robot a rigid recipe—good for simple tasks, but it failed for complex sentences like “The cat only purrs at night.”
- **1990s-2000s: Statistical NLP**: Computers learned patterns from data, like guessing the next word by studying tons of sentences. Think of a robot watching 1,000 chefs to learn cooking.
- **2010s-Now: Neural NLP**: Uses brain-like systems (neural networks) to understand context, like knowing “bank” means money or a river based on the sentence.

- **Analogy**: NLP’s journey is like a kid learning to talk: mimicking words, then understanding stories, now writing books.
- **Real-World Example**: In 2011, IBM Watson won *Jeopardy!* by understanding tricky questions. Now it helps scientists analyze cancer data.
- **Math Insight**: Claude Shannon’s entropy measures language unpredictability: H = -Σ p(x) log₂ p(x). Example: For a coin (p(heads)=0.5), H = -(0.5*log₂0.5 + 0.5*log₂0.5) = 1 bit. Low entropy = predictable text, helping NLG.

**Notebook Tip**: Write: “NLP evolved: Rules → Stats → Neural nets.” Ask: Why did rule-based NLP struggle with ambiguity?

**Exercise**: Calculate entropy for a biased coin (p(heads)=0.8). Solution: H = -(0.8*log₂0.8 + 0.2*log₂0.2) ≈ 0.72 bits.

````python
import math

# Entropy calculation for a biased coin
p_heads = 0.8
p_tails = 0.2
entropy = -(p_heads * math.log2(p_heads) + p_tails * math.log2(p_tails))
print(f"Entropy: {entropy:.2f} bits")
````

### 1.3 NLP’s Toolbox
NLP has tools like equipment in your lab:
- **Tokenization**: Breaks sentences into pieces (tokens). Example: “I love coding” → [“I”, “love”, “coding”].

- **Generation**: Creates text or code (our focus!).

- **Why for NL → Code?**: Tokenization splits instructions, embeddings understand meaning, and generation writes code.
- **Analogy**: NLP is a kitchen: tokenization chops veggies, embeddings taste flavors, generation serves the dish.

## Section 2: Natural Language Generation (NLG) - Creating Text

### 2.1 What is NLG?
NLG is when computers write human-like text from data, like turning “Rain, 20°C” into “It’s a rainy day with a high of 20°C.”

- **Simple Explanation**: NLG is a storyteller turning facts into sentences.
- **Link to NL → Code**: Instead of text, we generate code, like “Add two numbers” → `sum = a + b`.
- **Real-World Example**: News apps write sports summaries, like “Team A won 3-2.” In science, NLG turns lab data (e.g., “pH 7”) into reports.

**Notebook Tip**: Write: “NLG = Writing text from data.” Ask: How is code like a story?

### 2.2 How NLG Works
NLG predicts words one by one, like solving a puzzle:
- **Old Way (Pre-2017)**: RNNs read words slowly, one at a time.
- **New Way (Post-2017)**: Transformers read the whole sentence at once, like speed-reading.
- **Math**: NLG uses probabilities: P(sentence) = Π P(word_i | word_1...word_{i-1}). Example: P(“The cat sat”) ≈ P(“The”) * P(“cat”|“The”) * P(“sat”|“The cat”).

- **Analogy**: NLG is a librarian picking the best book to finish your story.

## Section 6: Attention Mechanism - The Key to Smarts

### 6.1 What’s Attention?
Attention lets the transformer focus on important words. For “Add two numbers,” it zooms in on “add” and “numbers” to write `+`.

- **Simple Explanation**: It’s like highlighting key notes in your textbook.
- **Analogy**: In a noisy classroom, attention tunes into the teacher’s voice.

### 6.2 Math of Attention
Formula: Attention(Q, K, V) = softmax(Q * K^T / sqrt(d_k)) * V
- Q (Query): What the AI’s looking for.
- K (Keys): Available words.
- V (Values): Info to use.
- d_k: Scaling factor.

**Example Calculation (2D)**:
- Words: “Add” (Q = [1, 0]), “two” (K1 = [1, 0], V1 = [2, 0]), “numbers” (K2 = [0, 1], V2 = [3, 1]).
- Step 1: Dot products: Q * K1^T = 1*1 + 0*0 = 1; Q * K2^T = 1*0 + 0*1 = 0.
- Step 2: Scale by sqrt(2) ≈ 1.41: 1/1.41 ≈ 0.71, 0/1.41 = 0.
- Step 3: Softmax: e^0.71 ≈ 2.03, e^0 = 1. Total = 3.03. Weights: 2.03/3.03 ≈ 0.67, 1/3.03 ≈ 0.33.
- Step 4: Output = 0.67*[2,0] + 0.33*[3,1] = [1.34, 0] + [0.99, 0.33] = [2.33, 0.33].

**What This Means**: AI focuses 67% on “two,” 33% on “numbers.”

**Exercise**: Recalculate with vectors [1,1], [0,1], [2,0]. Solution: Output ≈ [1.5, 0.5].

````python
import numpy as np

# Attention calculation
Q = np.array([1, 0])
K = np.array([[1, 0], [0, 1]])
V = np.array([[2, 0], [3, 1]])
d_k = 2

scores = np.dot(Q, K.T) / np.sqrt(d_k)
weights = np.exp(scores) / np.sum(np.exp(scores))
output = np.dot(weights, V)
print(f"Attention Output: {output}")
````

### 6.3 Types of Attention
- **Self-Attention**: Words in the input talk to each other.
- **Cross-Attention**: Code-writing part checks the instruction.
- **Multi-Head Attention**: Runs attention multiple times for different patterns.

**Analogy**: Multi-head is like multiple researchers studying the same problem from different angles.

## Section 11: Evaluation Metrics

- **Pass@k**: Chance at least one of k solutions works.
  - **Math**: P(pass@k) = 1 - C(n-k, n)/C(n, n). Example: 2/10 solutions work, pass@10 = 0.2.
- **BLEU**: Checks code similarity (less used).

**Real-World Example**: HumanEval tests 164 problems. AlphaCode excels at pass@100. <grok:render type="render_inline_citation">
<argument name="citation_id">30</argument>
</grok:render>

**Exercise**: Calculate pass@1 for 3/10 successes. Solution: 0.3.

````python
from math import comb

# Pass@k calculation
n, k, successes = 10, 1, 3
pass_k = 1 - comb(n - successes, k) / comb(n, k)
print(f"Pass@{k}: {pass_k}")
````

## Section 18: Major Project - Analyze a Dataset

Use NL → Code to generate a script for analyzing a real dataset (e.g., Iris dataset).

**Prompt**: “Analyze Iris dataset: calculate average petal length by species.”
**Steps**:
1. Load Iris dataset.
2. Group by species, compute mean petal length.
3. Plot results.

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Iris dataset
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target_names[iris.target]

# Calculate average petal length by species
avg_petal = df.groupby('species')['petal length (cm)'].mean()
print(avg_petal)

# Plot
sns.barplot(x=avg_petal.index, y=avg_petal.values)
plt.title('Average Petal Length by Species')
plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.show()
````
