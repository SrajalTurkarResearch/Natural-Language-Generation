## Section 2: Practical Code Guides
Step-by-step code for NLG and evaluation.

```python
from transformers import pipeline
import nltk
from nltk.translate.bleu_score import sentence_bleu
nltk.download('punkt')

# Simple NLG example using GPT-2
generator = pipeline('text-generation', model='gpt2')
generated = generator("The future of AI is", max_length=50)[0]['generated_text']
print(generated)
```

```python
# Quantitative: BLEU score
reference = [['This', 'is', 'a', 'test', 'sentence']]
candidate = ['This', 'is', 'test', 'sentence']
score = sentence_bleu(reference, candidate)
print(f'BLEU score: {score}')
```

## Section 6: Mini & Major Projects
### Mini Project: Simple NLG Evaluator
Generate text and compute BLEU.

```python
# Mini project code
prompt = 'Explain NLG'
gen_text = generator(prompt, max_length=100)[0]['generated_text']
ref = prompt + ' is the process of generating text from data.'
cand = nltk.word_tokenize(gen_text)
ref_tok = [nltk.word_tokenize(ref)]
print(sentence_bleu(ref_tok, cand))
```
