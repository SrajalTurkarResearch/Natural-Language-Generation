## Section 1: Theory & Tutorials – Understanding Logic-to-Text Mapping

### 1.1 What is NLG? The Big Picture
- **Definition**: NLG (Natural Language Generation) is when a computer turns facts, numbers, or rules into sentences that sound like a human wrote them. “Natural language” means everyday talking or writing, like English.
- **Analogy**: NLG is like a chef turning raw ingredients (facts) into a tasty dish (sentences).
- **Why for Scientists?**: Automates report writing, explains data, or simulates dialogues. Imagine Tesla using NLG to write instructions for his inventions.
- **History**: Started in the 1950s with Turing’s idea of talking machines. By the 2010s, deep learning (computers learning from examples) made NLG super smooth.
- **Subfields**:
  - Data-to-Text: Numbers to stories (e.g., sports scores to summaries).
  - Text-to-Text: Rewriting (e.g., summarizing).
  - Image-to-Text: Picture captions.
  - **Logic-to-Text**: Our focus – rules to sentences.

### 1.2 What is Logic-to-Text Mapping?
- **Definition**: Turning strict rules (logic, like “Eats(John, Apple)”) into normal sentences (“John eats an apple”). “Logic” is a clear, no-mistake way to write facts.
- **Why Cool?**:
  - **Control**: Change the rule, change the sentence.
  - **Truth**: Sticks to facts, no made-up stuff.
  - **Explain**: Show the rule to prove the sentence is right.
- **Analogy**: Logic is a recipe card; mapping is cooking the dish.

### 1.3 Types of Logic – The Building Blocks
1. **Propositional Logic**: True/false statements. Example: P = “It’s raining.” Q = “Take umbrella.” P → Q = “If raining, take umbrella.”
   - **Math**: Truth table:
     ```
     P | Q | P ∧ Q
     T | T | T
     T | F | F
     F | T | F
     F | F | F
     ```
2. **Predicate Logic**: About things and links. Example: ∀x (Human(x) → Mortal(x)) = “All humans are mortal.”
   - **Parts**: Predicates (e.g., Human(x)), Quantifiers (∀ = all, ∃ = some), Variables (x = any thing).
3. **Lambda Calculus**: Functions. Example: λx. Eats(John, x) = “John eats something.”
4. **Description Logic**: Knowledge maps. Example: Bird is an Animal with wings.
5. **Modal Logic**: Possibility/must. Example: ◇Rains = “Might rain.”

**Visualization**: Draw a tree for “All humans are mortal”:
```

|-- Human(x) → Mortal(x)
```

### 1.4 Mapping Process – Step-by-Step
1. **Parse Logic**: Check it’s correct.
2. **Content Selection**: Pick key facts.
3. **Discourse Planning**: Order ideas (e.g., cause then effect).
4. **Lexicalization**: Choose words (e.g., “Likes” → “enjoys”).
5. **Aggregation**: Combine facts (e.g., “John eats apple AND likes fruit” → one sentence).
6. **Referring Expressions**: Use “he” instead of “John” again.
7. **Surface Realization**: Add grammar.
8. **Evaluation**: Check with BLEU (word matches) or METEOR (similar words).

**Visualization**: Flowchart (draw this):
```
[Logic] → Parse → Select → Plan → Words → Combine → Pronouns → Grammar → [Text] → Check
```

**Easy Key Points**: NLG makes sentences from facts. Logic-to-text uses rules. Mapping has 8 steps.

## Section 2: Practical Code Guides – Build Your Own Mapper

Let’s code a rule-based logic-to-text mapper in Python. We’ll start simple, then add complexity.

### 2.1 Simple Rule-Based Mapper
Turn a single predicate into a sentence.

````python
def simple_logic_to_text(logic):
    # Logic is a dictionary, e.g., {'type': 'Predicate', 'subject': 'John', 'verb': 'eats', 'object': 'apple'}
    if logic['type'] == 'Predicate':
        return f"{logic['subject']} {logic['verb']} {logic['object']}."
    return "Unknown logic type."

# Test it
logic = {'type': 'Predicate', 'subject': 'John', 'verb': 'eats', 'object': 'apple'}
print(simple_logic_to_text(logic))  # Output: John eats apple.
````

**Explanation**:
- **Input**: A dictionary with logic parts (type, subject, verb, object).
- **Function**: Checks if it’s a predicate, then builds a sentence.
- **Why Simple?**: Like Turing’s basic code, it’s a starting point.

### 2.2 Advanced Mapper with AND
Handle combined logic (e.g., AND).

````python
def advanced_logic_to_text(logic):
    if logic['type'] == 'Predicate':
        return f"{logic['subject']} {logic['verb']} {logic['object']}."
    elif logic['type'] == 'AND':
        left_text = advanced_logic_to_text(logic['left'])
        right_text = advanced_logic_to_text(logic['right'])
        return f"{left_text[:-1]} and {right_text.lower()}"
    return "Unknown logic type."

# Test it
logic = {
    'type': 'AND',
    'left': {'type': 'Predicate', 'subject': 'John', 'verb': 'eats', 'object': 'apple'},
    'right': {'type': 'Predicate', 'subject': 'John', 'verb': 'likes', 'object': 'fruit'}
}
print(advanced_logic_to_text(logic))  # Output: John eats apple and likes fruit.
````

**Explanation**:
- **Recursive Function**: Handles nested logic (like a tree).
- **AND Logic**: Combines two predicates with “and.”

### 3.2 Mapping Process Flowchart
Use Matplotlib to sketch the process.

````python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch

fig, ax = plt.subplots(figsize=(10, 2))
ax.set_xlim(0, 5)
ax.set_ylim(0, 1)
steps = ['Logic', 'Parse', 'Select', 'Plan', 'Text']
for i, step in enumerate(steps):
    ax.text(i, 0.5, step, ha='center', va='center', bbox=dict(facecolor='lightblue', edgecolor='black'))
    if i < len(steps)-1:
        arrow = FancyArrowPatch((i+0.4, 0.5), (i+0.6, 0.5), mutation_scale=20)
        ax.add_patch(arrow)
ax.axis('off')
plt.show()
````

- **Flowchart**: Shows mapping steps like a conveyor belt.

- **Goal**: Map logic from the Logic2Text dataset.

````python
# Install datasets: pip install datasets
from datasets import load_dataset

dataset = load_dataset('logic2text', split='train[:1]')
print(dataset[0])  # Explore table and logic

# Simple rule for population
def map_population(logic, table):
    if 'Max(Population)' in logic:
        city = logic.split('=')[1].strip()
        pop = table[0]['Population']
        return f"{city} has the highest population, {pop}."

# Test (simplified example)
table = [{'City': 'New York', 'Population': '8M'}]
logic = 'Max(Population) = NewYork'
print(map_population(logic, table))
````
