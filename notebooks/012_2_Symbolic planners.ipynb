### 2.1 Simple Symbolic Planner Implementation
We'll implement a basic BFS planner in pure Python (no external libs needed beyond standard).

```python
from collections import deque

# Define states as dicts, actions as functions
def bfs_planner(initial_state, goal_state, actions):
    queue = deque([(initial_state, [])])
    visited = set()
    while queue:
        state, path = queue.popleft()
        if tuple(sorted(state.items())) in visited:
            continue
        visited.add(tuple(sorted(state.items())))
        if state == goal_state:
            return path
        for action in actions:
            if action'precondition':
                new_state = action'effect'
                queue.append((new_state, path + [action['name']]))
    return None

# Example: Blocksworld
initial = {'A': 'table', 'B': 'table'}
goal = {'A': 'B', 'B': 'table'}

actions = [
    {'name': 'stack A on B',
     'precondition': lambda s: s['A'] == 'table' and s['B'] == 'table',
     'effect': lambda s: {s, 'A': 'B'}}
]

plan = bfs_planner(initial, goal, actions)
print('Plan:', plan)
```

Explanation: Step 1: Define states. Step 2: BFS searches paths. From searches: Similar to PyReason but simplified.

### Exercise 1: Implement A Planner
Modify BFS to A. Solution below.

```python
import heapq

def a_star_planner(initial, goal, actions, heuristic):
    queue = [(0 + heuristic(initial), 0, initial, [])]
    visited = set()
    while queue:
        f, g, state, path = heapq.heappop(queue)
        if tuple(sorted(state.items())) in visited:
            continue
        visited.add(tuple(sorted(state.items())))
        if state == goal:
            return path
        for action in actions:
            if action'precondition':
                new_state = action'effect'
                new_g = g + 1
                heapq.heappush(queue, (new_g + heuristic(new_state), new_g, new_state, path + [action['name']]))
    return None

# Heuristic example
def heuristic(state):
    return 1 if state != goal else 0

plan = a_star_planner(initial, goal, actions, heuristic)
print('A* Plan:', plan)
```
