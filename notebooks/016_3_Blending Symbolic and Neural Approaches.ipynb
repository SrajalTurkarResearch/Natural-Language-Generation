## Section 1: Theory & Tutorials

### 1.1 What is NLG?
Natural Language Generation (NLG) is when computers create human-readable text from data, like turning numbers into sentences. Imagine a librarian who takes raw facts (like book records) and writes a story about them.

- **Key Idea**: NLG bridges data and human communication. It’s vital for science because it automates reports, explains models, or simulates hypotheses in words.
- **Example**: Input: Weather data (Temp: 75°F). Output: "It’s a warm, sunny day."
- **Analogy**: Like baking a cake – data (ingredients) becomes text (cake) through a process (recipe).

### 1.2 Symbolic AI in NLG
Symbolic AI uses clear rules and symbols (like words or concepts) to make text. It’s like a strict teacher following a textbook, precise but not flexible.

- **History**: Born in the 1950s with Turing’s ideas, used in systems like SHRDLU (1970) for describing block worlds.
- **How It Works**: Follows a pipeline: Plan content (what to say), structure sentences (how to say it), realize text (final words).
- **Logic**: Deterministic – if X, then Y. Example: If temp > 70°F, write "warm".
- **Pros**: Explainable, no huge data needed. **Cons**: Hard to scale, breaks if rules miss cases.
- **Math**: First-Order Logic (FOL). Example: ∀x (Cat(x) → Generate("feline")).

### 1.3 Neural AI in NLG
Neural AI learns patterns from data, like a child learning to talk by hearing others. It’s flexible but can make mistakes.

- **History**: Revived in 1986 with backpropagation, exploded with transformers (2017).
- **How It Works**: Encodes input (e.g., data to vectors), decodes to words using probabilities.
- **Math**: Softmax for word choice. P(word) = e^{score} / sum(e^{all scores}).
- **Pros**: Fluent, handles ambiguity. **Cons**: Needs lots of data, can “hallucinate” wrong facts.
- **Example**: Input: "Apples are red." Output: "Apples appear red in color."

### 1.4 Neuro-Symbolic AI in NLG
Mixing symbolic (rules) and neural (learning) creates systems that are accurate and fluent. It’s like a music conductor (symbolic) guiding a band (neural).

- **Why Mix?**: Neural for smooth text, symbolic for truth and structure. Solves neural’s “black box” problem for science.
- **Types (Kautz, 2022)**: Type 1 (neural with symbolic input), Type 2 (symbolic guides neural), etc.
- **Math**: Hybrid loss L = α L_neural + β L_symbolic, α + β = 1.
- **Example**: Neural drafts: "You have flu." Symbolic checks: If fever > 100°F, add "See doctor."
- **Real-World**: Medical reports, fact-checked news summaries.

## Section 2: Practical Code Guides

Let’s build a simple hybrid NLG system using Python. We’ll use NLTK for symbolic rules and Transformers for neural generation, with step-by-step explanations.

### 2.1 Symbolic NLG: Rule-Based Weather Report
We create a rule-based system to generate weather descriptions from data.

```python
def symbolic_weather_nlg(data):
    # Input: Dictionary with weather data
    temp = data.get('temp', 0)
    condition = data.get('condition', 'clear')
    
    # Rules for content planning
    if temp > 70:
        temp_desc = 'warm'
    elif temp < 50:
        temp_desc = 'cool'
    else:
        temp_desc = 'mild'
    
    # Sentence realization
    sentence = f"It's a {temp_desc}, {condition} day."
    return sentence

# Test
data = {'temp': 75, 'condition': 'sunny'}
print(symbolic_weather_nlg(data))  # Output: It's a warm, sunny day.
```

**Explanation**: This code uses simple if-then rules (symbolic) to decide words based on temperature and condition. It’s like a recipe: check the ingredients (data), follow steps (rules), get a dish (sentence). Try changing `temp` or `condition` in `data`.

```python
# Try it yourself
data2 = {'temp': 45, 'condition': 'cloudy'}
print(symbolic_weather_nlg(data2))  # Output: It's a cool, cloudy day.
```

### 2.2 Neural NLG: Transformer-Based Summary
We use a pre-trained transformer (e.g., T5) to summarize text.

```python
from transformers import pipeline

def neural_summary_nlg(text):
    summarizer = pipeline('summarization', model='t5-small')
    summary = summarizer(text, max_length=50, min_length=10, do_sample=False)
    return summary[0]['summary_text']

# Test
text = "Apples are red, juicy fruits that grow on trees and are rich in fiber."
print(neural_summary_nlg(text))  # Output: Apples are red, juicy, fiber-rich fruits grown on trees.
```

**Explanation**: The T5 model learns patterns from data to generate a summary. It’s like a student summarizing a book after reading many examples. The model encodes the input and decodes it into a shorter text. Try different `text` inputs.

### 2.3 Hybrid NLG: Combining Symbolic and Neural
We combine symbolic rules (for facts) with neural generation (for fluency).

```python
def hybrid_nlg(data, text):
    # Symbolic: Extract facts
    temp = data.get('temp', 0)
    if temp > 100:
        warning = 'Warning: Extreme heat! Stay hydrated.'
    else:
        warning = ''
    
    # Neural: Generate fluent description
    input_text = f"Weather data: temperature {temp}°F, condition {data.get('condition', 'clear')}."
    neural_text = neural_summary_nlg(input_text)
    
    # Combine
    return f"{neural_text} {warning}".strip()

# Test
data = {'temp': 105, 'condition': 'sunny'}
text = "It's very hot today with clear skies."
print(hybrid_nlg(data, text))  # Output: Sunny weather with high temperature. Warning: Extreme heat! Stay hydrated.
```

**Explanation**: The symbolic part checks facts (e.g., temp > 100°F), and the neural part makes the text smooth. It’s like a fact-checker and a writer working together. Try modifying the `data` dictionary.

## Section 3: Visualizations

Visuals help us understand complex systems. Let’s create a knowledge graph for symbolic NLG and a plot for neural probabilities.

### 3.1 Knowledge Graph for Symbolic NLG
We use NetworkX to visualize a simple knowledge graph.

```python
import networkx as nx
import matplotlib.pyplot as plt

def plot_knowledge_graph():
    G = nx.DiGraph()
    G.add_edges_from([('Weather', 'Temperature'), ('Weather', 'Condition'), 
                      ('Temperature', 'Warm'), ('Condition', 'Sunny')])
    plt.figure(figsize=(8, 6))
    pos = nx.spring_layout(G)
    nx.draw(G, pos, with_labels=True, node_color='lightblue', 
            node_size=2000, font_size=12, font_weight='bold')
    plt.title('Knowledge Graph for Weather NLG')
    plt.show()

plot_knowledge_graph()
```

**Explanation**: This graph shows relationships (e.g., Weather → Temperature → Warm). It’s like a family tree for data, guiding symbolic NLG rules. Sketch this in your notes.

### 3.2 Neural Probability Plot
We visualize word probabilities from a neural model.

```python
import numpy as np

def plot_word_probs():
    words = ['mat', 'dog', 'hat']
    logits = np.array([3.0, 1.0, 0.5])  # Example scores
    probs = np.exp(logits) / np.sum(np.exp(logits))  # Softmax
    plt.bar(words, probs, color='orange')
    plt.title('Neural Word Probabilities')
    plt.xlabel('Words')
    plt.ylabel('Probability')
    plt.show()

plot_word_probs()
```

**Explanation**: This bar plot shows the chance of picking each word (e.g., 'mat' is most likely). It’s like a voting system where neural picks the winner. Try different `logits` values.

### Exercise 1: Symbolic Rule Expansion
Add a new rule to `symbolic_weather_nlg` for wind speed (e.g., >20 mph = 'windy'). Test it.

**Solution**:

```python
def symbolic_weather_nlg_extended(data):
    temp = data.get('temp', 0)
    condition = data.get('condition', 'clear')
    wind = data.get('wind', 0)
    
    temp_desc = 'warm' if temp > 70 else 'cool' if temp < 50 else 'mild'
    wind_desc = 'windy' if wind > 20 else 'calm'
    
    return f"It's a {temp_desc}, {condition}, {wind_desc} day."

# Test
data = {'temp': 75, 'condition': 'sunny', 'wind': 25}
print(symbolic_weather_nlg_extended(data))  # Output: It's a warm, sunny, windy day.
```
