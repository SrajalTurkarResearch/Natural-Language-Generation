## 1. Fundamentals of Emotion Shaping in NLG

### 1.1 What is Emotion Shaping?
- **Definition:** Deliberate design of emotional trajectories (arcs) in generated narratives to evoke reader responses.
- **Why NLG?** Computers generate human-like stories for therapy, games, marketing.
- **Analogy (Einstein-style):** Emotions are "relativity" – relative to context, time, reader.

### 1.2 Core Emotions (Plutchik's Wheel)
| Emotion | Description | NLG Example |
|---------|-------------|-------------|
| Joy | Happiness | "She triumphed!" |
| Sadness | Loss | "Tears fell silently." |

### 1.3 Basic Emotional Arcs (Vonnegut/Reagan 2016)
- 6 Core Shapes: Rags to Riches, Tragedy, etc. (Visualized below).

```python
# Install if needed (run once)
# !pip install matplotlib plotly vaderSentiment textblob nltk transformers torch datasets huggingface_hub

import matplotlib.pyplot as plt
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

# Plot 6 Basic Emotional Arcs
def plot_emotional_arcs():
    fig = make_subplots(rows=2, cols=3, subplot_titles=['Rags to Riches', 'Tragedy', 'Man in Hole', 'Icarus', 'Cinderella', 'Oedipus'])
    x = np.linspace(0, 1, 100)

    # Rags to Riches
    fig.add_trace(go.Scatter(x=x, y=x, mode='lines', name='Rags to Riches'), row=1, col=1)
    # Tragedy
    fig.add_trace(go.Scatter(x=x, y=1-x, mode='lines', name='Tragedy'), row=1, col=2)
    # Man in Hole
    fig.add_trace(go.Scatter(x=x, y=np.sin(2*np.pi*x)*-0.5+0.5, mode='lines', name='Man in Hole'), row=1, col=3)
    # Icarus
    fig.add_trace(go.Scatter(x=x, y=np.sin(np.pi*x), mode='lines', name='Icarus'), row=2, col=1)
    # Cinderella
    fig.add_trace(go.Scatter(x=x, y=np.sin(2*np.pi*x), mode='lines', name='Cinderella'), row=2, col=2)
    # Oedipus
    fig.add_trace(go.Scatter(x=x, y=-np.sin(2*np.pi*x), mode='lines', name='Oedipus'), row=2, col=3)

    fig.update_layout(height=600, title_text="Six Basic Emotional Arcs (Reagan et al., 2016)")
    fig.show()

plot_emotional_arcs()
```

## 3. Mathematical Models

### 3.1 Quadratic/Cubic Arcs
$$ E(i) = a(i - \mu)^2 + b \quad (\text{Man in Hole})$$

### 3.2 Advanced: SVD Decomposition (Reagan 2016)

### 3.3 2025: Gaussian Processes for Dynamic Arcs

```python
from scipy.optimize import curve_fit
import numpy as np
import matplotlib.pyplot as plt

def quadratic(x, a, b, c):
    return a * x**2 + b * x + c

# Fit Arc to Sample Data
x_data = np.linspace(0, 1, 20)
y_data = np.sin(2 * np.pi * x_data) + np.random.normal(0, 0.05, 20)  # Noisy Cinderella

popt, _ = curve_fit(quadratic, x_data, y_data)

plt.plot(x_data, y_data, 'o', label='Data')
plt.plot(x_data, quadratic(x_data, *popt), '--', label='Fit')
plt.title('Fitting Emotional Arc (Cinderella)')
plt.legend()
plt.show()

print(f'Parameters: a={popt[0]:.3f}, b={popt[1]:.3f}, c={popt[2]:.3f}')
```

## 4. Practical NLG Implementation

### 4.1 Setup Lexicons

```python
import nltk
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
from textblob import TextBlob
nltk.download('vader_lexicon', quiet=True)

analyzer = SentimentIntensityAnalyzer()

def analyze_emotion(text):
    score = analyzer.polarity_scores(text)
    return score['compound']

# Example
text = "She felt overwhelming joy!"
print(f'Emotion Score: {analyze_emotion(text):.3f}')
```

### 4.2 Generate Story with Transformers

```python
from transformers import pipeline

generator = pipeline('text-generation', model='gpt2')

prompt = "In a lab, the scientist felt despair as the experiment failed, but then..."
story = generator(prompt, max_length=100, num_return_sequences=1)[0]['generated_text']
print(story)
print(f'Score: {analyze_emotion(story):.3f}')
```

### Exercise 1: Compute Arc
**Q:** Fit quadratic to [0.2, -0.5, 0.8].

**Solution:**

```python
# Solution Code
import numpy as np
from scipy.optimize import curve_fit

x = np.array([1, 2, 3])
y = np.array([0.2, -0.5, 0.8])

def quadratic(x, a, b, c):
    return a * x**2 + b * x + c

popt, _ = curve_fit(quadratic, x, y)
print(popt)
```
