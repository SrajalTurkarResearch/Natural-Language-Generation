## **1.1 THEORY: Tension Arcs Explained (Freytag's Pyramid)**

**Simple Definition**: A tension arc is the "excitement curve" of a story - low → high → resolution.

**5 Core Stages**:
1. **Exposition** (0% tension): "Once upon a time..."
2. **Rising Action** (25-75%): Problems build
3. **Climax** (100%): BIG moment!
4. **Falling Action** (75-25%): Resolution begins
5. **Denouement** (0%): "Happily ever after"

**Mathematical Model**:
$$T(t) = \begin{cases} 
a \cdot t^2 & 0 \leq t < 0.5 \\ 
-a \cdot (t-1)^2 + 1 & 0.5 \leq t \leq 1 
\end{cases}$$

Where $T(t)$ = tension, $t$ = story progress (0-1), $a$ = steepness

```python
# 🧮 INTERACTIVE TENSION ARC BUILDER
def plot_tension_arc(a=4, segments=5, title="Default Story"):
    t = np.linspace(0, 1, segments)
    tension = np.where(t < 0.5, a * t**2, -a * (t-1)**2 + 1)
    
    fig = go.Figure()
    fig.add_trace(go.Scatter(x=t, y=tension, mode='lines+markers', 
                            name='Tension', line=dict(color='red', width=4)))
    
    fig.update_layout(
        title=f"Tension Arc: {title}",
        xaxis_title="Story Progress", yaxis_title="Tension (0-1)",
        height=400, template='plotly_dark'
    )
    fig.show()
    return tension

# 🎛️ WIDGET CONTROL
a_slider = widgets.IntSlider(value=4, min=1, max=10, description="Steepness:")
seg_slider = widgets.IntSlider(value=5, min=3, max=10, description="Segments:")
title_text = widgets.Text(value="My Story", description="Title:")

widgets.interactive(plot_tension_arc, a=a_slider, segments=seg_slider, title=title_text)
print("🎯 ADJUST SLIDERS TO BUILD YOUR TENSION ARC!")
```

## **1.2 THEORY: Emotion Modeling Frameworks**

**3 SCIENTIFIC MODELS**:

| Model | Dimensions | Best For | Example |
|-------|------------|----------|---------|
| **Plutchik Wheel** | 8 Primary Emotions | Lexicon Building | Fear → Terror |
| **VAD (Russell)** | Valence, Arousal, Dominance | Math Modeling | Fear = (-0.8, 0.9, -0.7) |
| **Ekman** | 6 Universal | Cross-Cultural | Happy, Sad, Angry |

**VAD Formula**: $E = (V, A, D)$ where $-1 \leq V,A,D \leq 1$

**Emotion Distance**: $d(E_1, E_2) = \sqrt{(V_1-V_2)^2 + (A_1-A_2)^2 + (D_1-D_2)^2}$

```python
# 🧠 VAD EMOTION CALCULATOR & 3D VISUALIZER
class Emotion:
    def __init__(self, name, v, a, d):
        self.name = name
        self.v = v  # Valence
        self.a = a  # Arousal
        self.d = d  # Dominance
    
    def distance(self, other):
        return np.sqrt((self.v-other.v)**2 + (self.a-other.a)**2 + (self.d-other.d)**2)

# STANDARD EMOTIONS
emotions = {
    'Happy': Emotion('Happy', 0.8, 0.6, 0.6),
    'Sad': Emotion('Sad', -0.8, 0.2, -0.6),
    'Fear': Emotion('Fear', -0.7, 0.9, -0.8),
    'Anger': Emotion('Anger', -0.6, 0.8, 0.7),
    'Surprise': Emotion('Surprise', 0.2, 0.9, 0.0)
}

# 3D PLOT
fig = go.Figure(data=[go.Scatter3d(
    x=[e.v for e in emotions.values()],
    y=[e.a for e in emotions.values()],
    z=[e.d for e in emotions.values()],
    mode='markers+text',
    marker=dict(size=10, color='red'),
    text=list(emotions.keys()),
    textposition="middle center"
)])

fig.update_layout(title="3D VAD Emotion Space", scene=dict(
    xaxis_title="Valence", yaxis_title="Arousal", zaxis_title="Dominance"
))
fig.show()

# DISTANCE CALCULATOR
print("📏 EMOTION DISTANCES:")
for e1 in emotions:
    for e2 in emotions:
        if e1 != e2:
            dist = emotions[e1].distance(emotions[e2])
            print(f"{e1} → {e2}: {dist:.2f}")
```

## **2.1 CODE GUIDE: Rule-Based NLG System**

**Step-by-Step**: Build a complete story generator in 47 lines!

1. **Lexicon**: Emotion → Words
2. **Tension Mapper**: Level → Sentence Style
3. **Arc Generator**: Combine both
4. **Output**: Complete story

```python
# 🏗️ COMPLETE RULE-BASED NLG SYSTEM
class EmotionNLG:
    def __init__(self):
        # 1. EMOTION LEXICON
        self.lexicon = {
            'curious': ['wondered', 'explored', 'discovered'],
            'nervous': ['hesitated', 'glanced nervously', 'felt uneasy'],
            'scared': ['panicked', 'froze', 'heart raced'],
            'terrified': ['screamed', 'ran', 'horrified'],
            'relieved': ['sighed', 'smiled', 'thanked']
        }
        
        # 2. TENSION STYLES
        self.styles = {
            0: lambda x: f"The {x} was peaceful.",
            3: lambda x: f"Suddenly, the {x} seemed strange.",
            7: lambda x: f"The {x} was terrifying!",
            10: lambda x: f"'{x.upper()}!' they screamed!!!",
            2: lambda x: f"Finally, the {x} was safe again."
        }
    
    def generate_story(self, arc=[0,3,7,10,2], theme='forest'):
        emotions = ['curious', 'nervous', 'scared', 'terrified', 'relieved']
        story = []
        
        for i, tension in enumerate(arc):
            emotion = emotions[i]
            verb = np.random.choice(self.lexicon[emotion])
            sentence = self.styles[tension](f"team {verb} the {theme}")
            story.append(f"{i+1}. {sentence}")
        
        return '\n'.join(story)

# 🧪 TEST IT
nlg = EmotionNLG()
print("📖 GENERATED STORY:")
print(nlg.generate_story())

# 🎛️ INTERACTIVE VERSION
theme_widget = widgets.Text(value='spaceship')
widgets.interactive(lambda t: print(nlg.generate_story(theme=t)), t=theme_widget)
```

## **3.1 APPLICATION: Video Game NPC Dialogue**

**Real Case**: *The Last of Us* style dynamic dialogue

**Flow**:
1. Game state → Tension level
2. Tension → Emotion
3. Emotion → Dialogue line
4. Output → Voice synthesis

```python
# 🎮 VIDEO GAME NPC SYSTEM (COMPLETE)
class GameNPC:
    def __init__(self):
        self.dialogue_bank = {
            (0, 'calm'): "The weather's nice today...",
            (3, 'nervous'): "Did you hear that noise?",
            (7, 'scared'): "Something's out there! Get ready!",
            (10, 'panic'): "RUN! IT'S COMING RIGHT NOW!!!",
            (2, 'relief'): "Whew... I thought we were done for."
        }
    
    def speak(self, tension, context='combat'):
        emotion = {0:'calm', 3:'nervous', 7:'scared', 10:'panic', 2:'relief'}[tension]
        line = self.dialogue_bank[(tension, emotion)]
        return f"🎙️ NPC: {line}"

# 🕹️ GAME SIMULATION
npc = GameNPC()
game_arc = [0, 3, 7, 10, 2]

print("=== GAMEPLAY DIALOGUE ===")
for step, tension in enumerate(game_arc):
    print(f"Step {step+1} (Tension {tension}): {npc.speak(tension)}")
    
# 📊 VISUALIZE GAME ARC
fig = px.line(x=range(5), y=game_arc, title="Game Tension Arc")
fig.show()
```

**Chen's Multi-Arc Formula**:
$$R = \sum_{i=1}^n w_i \cdot T_i(t) \cdot E_i(t)$$

Where $R$ = Total Resonance, $T_i$ = Tension Arc i, $E_i$ = Emotion i, $w_i$ = Weight
