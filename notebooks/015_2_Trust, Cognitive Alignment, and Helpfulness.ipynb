Explanation: This generates NLG output. To evaluate trust, check accuracy manually or with fact-checkers.

```python
# Trust Score Calculation
def trust_score(acc, trans, rel):
    """Compute average trust score."""
    vals = [acc, trans, rel]
    if not all(isinstance(x, (int, float)) for x in vals):
        raise ValueError("All inputs must be numeric.")
    if not all(0.0 <= x <= 1.0 for x in vals):
        raise ValueError("All inputs must be in [0, 1].")
    return sum(vals) / 3
print(trust_score(0.8, 0.6, 0.9))  # 0.7666...
```

## Mini Projects

Mini: Evaluate LLM output for helpfulness using BLEU.

```python
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
reference = [['The', 'quick', 'brown', 'fox', 'jumps']]
candidate = ['Quick', 'brown', 'fox', 'jumps', 'over']
# Lowercase everything for proper BLEU calculation
reference = [[token.lower() for token in sent] for sent in reference]
candidate = [token.lower() for token in candidate]
smooth = SmoothingFunction().method1
score = sentence_bleu(reference, candidate, smoothing_function=smooth)
print(score)
```

## Exercises

Exercise 1: Compute JSD for alignment.

Solution:

```python
import numpy as np
from scipy.spatial.distance import jensenshannon
p = np.array([0.6, 0.4])
q = np.array([0.5, 0.5])
# JSD is always >= 0
jsd = jensenshannon(p, q) ** 2
print(jsd)  # Approx 0.005
```
